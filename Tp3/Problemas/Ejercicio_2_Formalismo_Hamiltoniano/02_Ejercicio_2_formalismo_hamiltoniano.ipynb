# Ejercicio 2: Formalismo Hamiltoniano

Partimos del Hamiltoniano de interacción a cuatro ondas

$$
\mathcal H
=
\int \omega_k |a_k|^2\,dk
+
\int
g\,a_k^*a_1^*a_2a_3\,
\delta(k+k_1-k_2-k_3)\,
dk_1\,dk_2\,dk_3\,dk.
$$

Aquí $g$ es una amplitud de scattering constante, $a_k$ es la amplitud compleja del modo $k$ y $\omega_k$ es la relación de dispersión.

Del Ejercicio 1, en el régimen considerado, la dispersión puede escribirse como

$$
\omega(k)
\simeq
c_0k-\alpha k^2.
$$

El término lineal $c_0k$ puede eliminarse mediante una transformación de Galileo, es decir, pasando a un sistema que se mueve con velocidad $c_0$. Siguiendo la convención indicada en la guía y en la errata, tomamos para lo que sigue

$$
\omega_k=\alpha k^2,
\qquad
\alpha>0.
$$

Con esta elección obtendremos la ecuación de Schrödinger no lineal

$$
i\frac{\partial a}{\partial t}
=
-\alpha\frac{\partial^2 a}{\partial x^2}
+g|a|^2a.
$$

## (a) Ecuación diferencial para $\dot a_k$

Las ecuaciones de Hamilton para las amplitudes complejas son

$$
i\dot a_k
=
\frac{\delta \mathcal H}{\delta a_k^*}.
$$

Separamos el Hamiltoniano en parte lineal y parte no lineal:

$$
\mathcal H=\mathcal H_L+\mathcal H_{NL},
$$

con

$$
\mathcal H_L
=
\int \omega_q |a_q|^2\,dq
=
\int \omega_q a_q^*a_q\,dq,
$$

y

$$
\mathcal H_{NL}
=
\int
g\,a_q^*a_1^*a_2a_3
\delta(q+k_1-k_2-k_3)
\,dk_1\,dk_2\,dk_3\,dq.
$$

Primero calculamos la variación funcional de la parte lineal:

$$
\frac{\delta \mathcal H_L}{\delta a_p^*}
=
\frac{\delta}{\delta a_p^*}
\int \omega_q a_q^*a_q\,dq.
$$

Usamos

$$
\frac{\delta a_q^*}{\delta a_p^*}
=
\delta(q-p).
$$

Entonces

$$
\frac{\delta \mathcal H_L}{\delta a_p^*}
=
\int \omega_q a_q\,\delta(q-p)\,dq.
$$

La delta evalúa $q=p$, por lo que

$$
\boxed{
\frac{\delta \mathcal H_L}{\delta a_p^*}
=
\omega_p a_p
}.
$$

Ahora calculamos la variación de la parte no lineal. Debemos tener cuidado porque $a_p^*$ puede aparecer en dos lugares: como $a_q^*$ o como $a_1^*$.

Partimos de

$$
\mathcal H_{NL}
=
\int
g\,a_q^*a_1^*a_2a_3
\delta(q+k_1-k_2-k_3)
\,dk_1\,dk_2\,dk_3\,dq.
$$

Derivando respecto de $a_p^*$,

$$
\frac{\delta}{\delta a_p^*}
\left(
a_q^*a_1^*a_2a_3
\right)
=
\delta(q-p)a_1^*a_2a_3
+
a_q^*\delta(k_1-p)a_2a_3.
$$

Por lo tanto,

$$
\frac{\delta \mathcal H_{NL}}{\delta a_p^*}
=
I_1+I_2,
$$

donde

$$
I_1
=
g\int
\delta(q-p)a_1^*a_2a_3
\delta(q+k_1-k_2-k_3)
\,dk_1\,dk_2\,dk_3\,dq,
$$

y

$$
I_2
=
g\int
a_q^*\delta(k_1-p)a_2a_3
\delta(q+k_1-k_2-k_3)
\,dk_1\,dk_2\,dk_3\,dq.
$$

En el primer término, la delta $\delta(q-p)$ fija $q=p$:

$$
I_1
=
g\int
a_1^*a_2a_3
\delta(p+k_1-k_2-k_3)
\,dk_1\,dk_2\,dk_3.
$$

En el segundo término, $\delta(k_1-p)$ fija $k_1=p$:

$$
I_2
=
g\int
a_q^*a_2a_3
\delta(q+p-k_2-k_3)
\,dq\,dk_2\,dk_3.
$$

Renombramos la variable muda $q$ como $k_1$. Entonces

$$
I_2
=
g\int
a_1^*a_2a_3
\delta(k_1+p-k_2-k_3)
\,dk_1\,dk_2\,dk_3.
$$

Como

$$
\delta(k_1+p-k_2-k_3)
=
\delta(p+k_1-k_2-k_3),
$$

tenemos

$$
I_2=I_1.
$$

Luego,

$$
\frac{\delta \mathcal H_{NL}}{\delta a_p^*}
=
2g\int
a_1^*a_2a_3
\delta(p+k_1-k_2-k_3)
\,dk_1\,dk_2\,dk_3.
$$

Por lo tanto, la ecuación de Hamilton queda

$$
i\dot a_p
=
\omega_p a_p
+
2g\int
a_1^*a_2a_3
\delta(p+k_1-k_2-k_3)
\,dk_1\,dk_2\,dk_3.
$$

Renombrando $p\to k$,

$$
\boxed{
i\dot a_k
=
\omega_k a_k
+
2g\int
a_1^*a_2a_3
\delta(k+k_1-k_2-k_3)
\,dk_1\,dk_2\,dk_3
}.
$$

Como $g$ es una constante de scattering, el factor $2$ puede absorberse redefiniendo $g$. Con esa convención,

$$
\boxed{
i\dot a_k
=
\omega_k a_k
+
g\int
a_1^*a_2a_3
\delta(k+k_1-k_2-k_3)
\,dk_1\,dk_2\,dk_3
}.
$$

## (b) Ecuación diferencial en el espacio físico

Usamos la transformada de Fourier

$$
a(x,t)
=
\int a_k(t)e^{ikx}\,dk.
$$

La ecuación modal, absorbiendo el factor $2$ en la definición de $g$, es

$$
i\dot a_k
=
\omega_k a_k
+
g\int
a_1^*a_2a_3
\delta(k+k_1-k_2-k_3)
\,dk_1\,dk_2\,dk_3.
$$

Tomamos

$$
\omega_k=\alpha k^2.
$$

La parte lineal en espacio físico se obtiene multiplicando por $e^{ikx}$ e integrando en $k$:

$$
\int \alpha k^2 a_k e^{ikx}\,dk.
$$

Como

$$
\frac{\partial^2 a}{\partial x^2}
=
\int -k^2 a_k e^{ikx}\,dk,
$$

entonces

$$
\int \alpha k^2 a_k e^{ikx}\,dk
=
-\alpha\frac{\partial^2 a}{\partial x^2}.
$$

Ahora transformamos la parte no lineal. Consideramos

$$
\int
\left[
\int
g\,a_1^*a_2a_3
\delta(k+k_1-k_2-k_3)
\,dk_1\,dk_2\,dk_3
\right]
e^{ikx}\,dk.
$$

La delta impone

$$
k=-k_1+k_2+k_3.
$$

Entonces

$$
e^{ikx}
=
e^{-ik_1x}e^{ik_2x}e^{ik_3x}.
$$

Por lo tanto,

$$
g\int
a_1^*a_2a_3
e^{-ik_1x}e^{ik_2x}e^{ik_3x}
\,dk_1\,dk_2\,dk_3.
$$

Separando los factores,

$$
g
\left(
\int a_1^*e^{-ik_1x}\,dk_1
\right)
\left(
\int a_2e^{ik_2x}\,dk_2
\right)
\left(
\int a_3e^{ik_3x}\,dk_3
\right).
$$

Reconocemos

$$
\int a_1^*e^{-ik_1x}\,dk_1
=
a^*(x,t),
$$

$$
\int a_2e^{ik_2x}\,dk_2
=
a(x,t),
$$

$$
\int a_3e^{ik_3x}\,dk_3
=
a(x,t).
$$

Entonces la parte no lineal es

$$
g\,a^*(x,t)a(x,t)a(x,t)
=
g|a(x,t)|^2a(x,t).
$$

Juntando la parte lineal y la no lineal, obtenemos

$$
\boxed{
i\frac{\partial a}{\partial t}
=
-\alpha\frac{\partial^2 a}{\partial x^2}
+g|a|^2a
}.
$$

En más dimensiones, la derivada segunda se reemplaza por el laplaciano:

$$
\boxed{
i\frac{\partial a}{\partial t}
=
-\alpha\nabla^2 a
+g|a|^2a
}.
$$

Esta ecuación no representa, en general, un campo escalar real. La razón es que $a$ es una amplitud compleja. Si $a$ fuese real, entonces el lado izquierdo

$$
i\frac{\partial a}{\partial t}
$$

sería puramente imaginario, mientras que el lado derecho

$$
-\alpha\nabla^2a+g|a|^2a
$$

sería real. Para que ambos coincidan, salvo casos estacionarios o triviales, se necesita que $a$ sea complejo.

## (c) Conservación de $E(t)$

Queremos mostrar que

$$
E(t)
=
\int |a(x,t)|^2\,dx
$$

es constante en el tiempo bajo condiciones de borde adecuadas.

Derivamos:

$$
\frac{dE}{dt}
=
\frac{d}{dt}
\int a a^*\,dx.
$$

Entonces

$$
\frac{dE}{dt}
=
\int
\left(
a_t a^*
+
a\,a_t^*
\right)dx.
$$

La ecuación de evolución es

$$
i a_t
=
-\alpha a_{xx}
+g|a|^2a.
$$

Despejamos $a_t$:

$$
a_t
=
i\alpha a_{xx}
-ig|a|^2a.
$$

Conjugando,

$$
a_t^*
=
-i\alpha a_{xx}^*
+ig|a|^2a^*.
$$

Sustituimos en la derivada de $E$:

$$
\frac{dE}{dt}
=
\int
\left[
\left(
i\alpha a_{xx}
-ig|a|^2a
\right)a^*
+
a
\left(
-i\alpha a_{xx}^*
+ig|a|^2a^*
\right)
\right]dx.
$$

Separamos los términos:

$$
\frac{dE}{dt}
=
i\alpha
\int
\left(
a_{xx}a^*
-
a a_{xx}^*
\right)dx
+
\int
\left(
-ig|a|^2aa^*
+ig|a|^2aa^*
\right)dx.
$$

Los términos no lineales se cancelan exactamente:

$$
-ig|a|^2aa^*
+ig|a|^2aa^*
=0.
$$

Entonces

$$
\frac{dE}{dt}
=
i\alpha
\int
\left(
a_{xx}a^*
-
a a_{xx}^*
\right)dx.
$$

Integramos por partes. Para el primer término,

$$
\int a_{xx}a^*\,dx
=
\left[
a_xa^*
\right]_{\partial D}
-
\int a_xa_x^*\,dx.
$$

Para el segundo,

$$
\int a a_{xx}^*\,dx
=
\left[
aa_x^*
\right]_{\partial D}
-
\int a_xa_x^*\,dx.
$$

Restando,

$$
\int
\left(
a_{xx}a^*
-
a a_{xx}^*
\right)dx
=
\left[
a_xa^*
-aa_x^*
\right]_{\partial D}.
$$

Por lo tanto,

$$
\frac{dE}{dt}
=
i\alpha
\left[
a_xa^*
-aa_x^*
\right]_{\partial D}.
$$

Luego $E$ se conserva si el término de borde se anula:

$$
\left[
a_xa^*
-aa_x^*
\right]_{\partial D}=0.
$$

Esto ocurre, por ejemplo, en dominios periódicos, en la recta si $a$ y sus derivadas decaen suficientemente rápido, o con condiciones de borde que anulen el flujo de norma.

## (d) Soluciones tipo onda viajera

Proponemos

$$
a(x,t)
=
\phi(s)e^{i\xi},
$$

donde

$$
s=x-vt,
$$

y

$$
\xi=\kappa x-\Omega t.
$$

Supondremos que $\phi(s)$ es real. Calculamos las derivadas separando cuidadosamente la dependencia en $s$ y en $\xi$.

Primero,

$$
\frac{\partial s}{\partial t}=-v,
\qquad
\frac{\partial \xi}{\partial t}=-\Omega.
$$

Entonces

$$
\frac{\partial a}{\partial t}
=
\left(
-v\frac{d\phi}{ds}
-i\Omega\phi
\right)e^{i\xi}.
$$

Luego,

$$
\frac{\partial s}{\partial x}=1,
\qquad
\frac{\partial \xi}{\partial x}=\kappa.
$$

Por lo tanto,

$$
\frac{\partial a}{\partial x}
=
\left(
\frac{d\phi}{ds}
+i\kappa\phi
\right)e^{i\xi}.
$$

Derivando una vez más,

$$
\frac{\partial^2 a}{\partial x^2}
=
\left(
\frac{d^2\phi}{ds^2}
+2i\kappa\frac{d\phi}{ds}
-\kappa^2\phi
\right)e^{i\xi}.
$$

Sustituimos en la ecuación

$$
i a_t
=
-\alpha a_{xx}
+g|a|^2a.
$$

El lado izquierdo es

$$
i a_t
=
\left(
-iv\frac{d\phi}{ds}
+\Omega\phi
\right)e^{i\xi}.
$$

El lado derecho es

$$
-\alpha a_{xx}
+g|a|^2a
=
\left[
-\alpha\frac{d^2\phi}{ds^2}
-2i\alpha\kappa\frac{d\phi}{ds}
+\alpha\kappa^2\phi
+g\phi^3
\right]e^{i\xi}.
$$

Igualamos partes imaginarias:

$$
-v\frac{d\phi}{ds}
=
-2\alpha\kappa\frac{d\phi}{ds}.
$$

Para una solución no constante, $d\phi/ds$ no es idénticamente cero, de modo que

$$
\boxed{
v=2\alpha\kappa
}.
$$

Ahora igualamos partes reales:

$$
\Omega\phi
=
-\alpha\frac{d^2\phi}{ds^2}
+\alpha\kappa^2\phi
+g\phi^3.
$$

Reordenando,

$$
\boxed{
-\alpha\frac{d^2\phi}{ds^2}
+
\lambda\phi
+
g\phi^3
=0
}
$$

donde definimos

$$
\lambda=\alpha\kappa^2-\Omega.
$$

### Solitones brillantes: $g<0$

Tomamos

$$
g=-|g|.
$$

La ecuación para $\phi$ queda

$$
-\alpha\frac{d^2\phi}{ds^2}
+
\lambda\phi
-
|g|\phi^3
=0.
$$

Buscamos una solución localizada, que tienda a cero lejos del centro:

$$
\phi(s)\to 0
\qquad
\text{cuando}
\qquad
|s|\to\infty.
$$

Multiplicamos la ecuación por $d\phi/ds$:

$$
-\alpha\frac{d^2\phi}{ds^2}\frac{d\phi}{ds}
+
\lambda\phi\frac{d\phi}{ds}
-
|g|\phi^3\frac{d\phi}{ds}
=0.
$$

Reconocemos derivadas totales:

$$
\frac{d}{ds}
\left[
-\frac{\alpha}{2}
\left(
\frac{d\phi}{ds}
\right)^2
+
\frac{\lambda}{2}\phi^2
-
\frac{|g|}{4}\phi^4
\right]
=0.
$$

Por lo tanto,

$$
-\frac{\alpha}{2}
\left(
\frac{d\phi}{ds}
\right)^2
+
\frac{\lambda}{2}\phi^2
-
\frac{|g|}{4}\phi^4
=C.
$$

Como para $|s|\to\infty$ se tiene $\phi\to0$ y $d\phi/ds\to0$, resulta $C=0$.

Entonces

$$
\left(
\frac{d\phi}{ds}
\right)^2
=
\frac{\lambda}{\alpha}\phi^2
\left(
1-\frac{|g|}{2\lambda}\phi^2
\right).
$$

Definimos la amplitud

$$
A^2=\frac{2\lambda}{|g|}.
$$

Así,

$$
\left(
\frac{d\phi}{ds}
\right)^2
=
\frac{\lambda}{\alpha}\phi^2
\left(
1-\frac{\phi^2}{A^2}
\right).
$$

Esta ecuación se integra separando variables:

$$
\frac{d\phi}
{\phi\sqrt{1-\phi^2/A^2}}
=
\pm
\sqrt{\frac{\lambda}{\alpha}}\,ds.
$$

La solución localizada correspondiente es

$$
\boxed{
\phi(s)
=
A\,\operatorname{sech}
\left[
B(s-s_0)
\right]
}
$$

con

$$
B=\sqrt{\frac{\lambda}{\alpha}},
\qquad
A=\sqrt{\frac{2\lambda}{|g|}}.
$$

Como $g<0$, también podemos escribir

$$
A=\sqrt{\frac{-2\lambda}{g}}.
$$

Para que $A$ y $B$ sean reales, necesitamos

$$
\lambda>0.
$$

Recordando que

$$
\lambda=\alpha\kappa^2-\Omega,
$$

obtenemos

$$
\Omega=\alpha\kappa^2-\lambda.
$$

La solución brillante completa es

$$
\boxed{
a(x,t)
=
A\,\operatorname{sech}
\left[
B(x-vt-s_0)
\right]
\exp\left(i(\kappa x-\Omega t)\right)
}
$$

con

$$
v=2\alpha\kappa,
\qquad
B=\sqrt{\frac{\lambda}{\alpha}},
\qquad
A=\sqrt{\frac{-2\lambda}{g}},
\qquad
\Omega=\alpha\kappa^2-\lambda.
$$

Se llama solitón brillante porque el módulo $|a|=|\phi|$ es un pulso localizado sobre un fondo nulo.

### Solitones oscuros: $g>0$

Ahora tomamos

$$
g>0.
$$

La ecuación de perfil es

$$
-\alpha\frac{d^2\phi}{ds^2}
+
\lambda\phi
+
g\phi^3
=0.
$$

Los solitones oscuros no viven sobre fondo cero. En cambio, buscamos soluciones que tiendan a amplitudes constantes opuestas:

$$
\phi(s)\to \pm A
\qquad
\text{cuando}
\qquad
s\to\pm\infty.
$$

Para que $\phi=A$ sea una solución constante de la ecuación de perfil, debe cumplirse

$$
\lambda A+gA^3=0.
$$

Si $A\neq0$,

$$
\lambda=-gA^2.
$$

Como $g>0$, esto implica

$$
\lambda<0.
$$

Integramos una vez la ecuación de perfil. Multiplicamos por $d\phi/ds$:

$$
-\alpha\frac{d^2\phi}{ds^2}\frac{d\phi}{ds}
+
\lambda\phi\frac{d\phi}{ds}
+
g\phi^3\frac{d\phi}{ds}
=0.
$$

Esto da

$$
\frac{d}{ds}
\left[
-\frac{\alpha}{2}
\left(
\frac{d\phi}{ds}
\right)^2
+
\frac{\lambda}{2}\phi^2
+
\frac{g}{4}\phi^4
\right]
=0.
$$

Luego,

$$
-\frac{\alpha}{2}
\left(
\frac{d\phi}{ds}
\right)^2
+
\frac{\lambda}{2}\phi^2
+
\frac{g}{4}\phi^4
=C.
$$

Para un solitón oscuro,

$$
\phi\to \pm A,
\qquad
\frac{d\phi}{ds}\to0
\qquad
\text{cuando}
\qquad
|s|\to\infty.
$$

Por lo tanto,

$$
C=
\frac{\lambda}{2}A^2
+
\frac{g}{4}A^4.
$$

Usando $\lambda=-gA^2$,

$$
C=
-\frac{g}{2}A^4
+
\frac{g}{4}A^4
=
-\frac{g}{4}A^4.
$$

Sustituimos $C$:

$$
-\frac{\alpha}{2}
\left(
\frac{d\phi}{ds}
\right)^2
+
\frac{\lambda}{2}\phi^2
+
\frac{g}{4}\phi^4
=
-\frac{g}{4}A^4.
$$

Multiplicando por $-2/\alpha$,

$$
\left(
\frac{d\phi}{ds}
\right)^2
=
\frac{g}{2\alpha}
\left(
A^2-\phi^2
\right)^2.
$$

Tomando raíz,

$$
\frac{d\phi}{ds}
=
\pm
\sqrt{\frac{g}{2\alpha}}
\left(
A^2-\phi^2
\right).
$$

Separamos variables:

$$
\frac{d\phi}{A^2-\phi^2}
=
\pm
\sqrt{\frac{g}{2\alpha}}\,ds.
$$

La integral es

$$
\int
\frac{d\phi}{A^2-\phi^2}
=
\frac{1}{A}
\operatorname{arctanh}
\left(
\frac{\phi}{A}
\right).
$$

Entonces

$$
\operatorname{arctanh}
\left(
\frac{\phi}{A}
\right)
=
\pm
A\sqrt{\frac{g}{2\alpha}}
(s-s_0).
$$

Aplicando tangente hiperbólica,

$$
\boxed{
\phi(s)
=
A\,\tanh
\left[
B(s-s_0)
\right]
}
$$

con

$$
B=A\sqrt{\frac{g}{2\alpha}}.
$$

Como $\lambda=-gA^2$, también

$$
B=\sqrt{\frac{-\lambda}{2\alpha}}.
$$

La solución oscura completa es

$$
\boxed{
a(x,t)
=
A\,\tanh
\left[
B(x-vt-s_0)
\right]
\exp\left(i(\kappa x-\Omega t)\right)
}
$$

con

$$
v=2\alpha\kappa,
$$

$$
B=A\sqrt{\frac{g}{2\alpha}},
$$

y como

$$
\lambda=\alpha\kappa^2-\Omega=-gA^2,
$$

resulta

$$
\boxed{
\Omega=\alpha\kappa^2+gA^2
}.
$$

Se llama solitón oscuro porque el módulo presenta una depresión localizada sobre un fondo no nulo. A diferencia del solitón brillante, no es un pulso sobre fondo cero, sino una “muesca” o defecto de amplitud.

### Solución de amplitud constante

También podemos tomar

$$
\phi(s)=A,
$$

constante. Entonces

$$
\frac{d^2\phi}{ds^2}=0.
$$

La ecuación de perfil

$$
-\alpha\frac{d^2\phi}{ds^2}
+
\lambda\phi
+
g\phi^3
=0
$$

queda

$$
\lambda A+gA^3=0.
$$

Si $A\neq0$,

$$
\lambda=-gA^2.
$$

Como

$$
\lambda=\alpha\kappa^2-\Omega,
$$

obtenemos

$$
\boxed{
\Omega=\alpha\kappa^2+gA^2
}.
$$

Entonces

$$
\boxed{
a(x,t)
=
A e^{i(\kappa x-\Omega t)}
}
$$

es solución. Esto muestra que una onda monocromática se mantiene monocromática: la no linealidad solo modifica su frecuencia mediante el corrimiento

$$
gA^2.
$$

### Gráficos de los perfiles $\phi(s)$

Para visualizar la diferencia entre los dos tipos de solitones, graficamos perfiles normalizados:

$$
\phi_{\text{brillante}}(s)=A\,\operatorname{sech}(Bs),

$$
\phi_{\text{oscuro}}(s)=A\,\tanh(Bs).
$$

```python
import numpy as np
import matplotlib.pyplot as plt

s = np.linspace(-8, 8, 800)
A = 1.0
B = 1.0

phi_bright = A / np.cosh(B*s)
phi_dark = A * np.tanh(B*s)

fig, axes = plt.subplots(1, 2, figsize=(12, 4), constrained_layout=True)

axes[0].plot(s, phi_bright, lw=2, color="tab:orange")
axes[0].set_title("Solitón brillante")
axes[0].set_xlabel(r"$s=x-vt$")
axes[0].set_ylabel(r"$\phi(s)$")
axes[0].grid(True, alpha=0.3)

axes[1].plot(s, phi_dark, lw=2, color="tab:blue")
axes[1].axhline(A, color="k", ls="--", lw=1, alpha=0.6)
axes[1].axhline(-A, color="k", ls="--", lw=1, alpha=0.6)
axes[1].set_title("Solitón oscuro")
axes[1].set_xlabel(r"$s=x-vt$")
axes[1].set_ylabel(r"$\phi(s)$")
axes[1].grid(True, alpha=0.3)

plt.show()
```

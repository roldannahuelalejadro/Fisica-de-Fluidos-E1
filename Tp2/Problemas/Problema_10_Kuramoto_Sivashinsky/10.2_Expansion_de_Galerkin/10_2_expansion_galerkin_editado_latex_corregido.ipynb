# Problema 10.2 — Expansión de Galerkin y bifurcaciones

En esta sección se proyecta la ecuación de Kuramoto–Sivashinsky reescalada,

$$
u_t=-u_{xx}-k^2u_{xxxx}-u u_x,
\qquad
k=\frac{2\pi}{L},
$$

sobre el subespacio impar generado por senos. Primero se obtiene el sistema modal infinito, luego se estudia la estabilidad del equilibrio trivial y finalmente se trunca a dos modos para analizar puntos fijos y bifurcaciones en función del tamaño del dominio.

| Inciso | Objetivo |
|---|---|
| 10.2(i) | Derivar el sistema modal mediante proyección de Galerkin |
| 10.2(ii) | Linealizar alrededor de $a_\ell=0$ y determinar cuándo aparecen modos inestables |
| 10.2(iii) | Truncar el sistema a $a_1,a_2$ |
| 10.2(iv) | Calcular puntos fijos y estabilidad del sistema truncado |
| 10.2(v) | Construir el diagrama de bifurcaciones en función de $L$ |

La notación que usaremos es: $\ell$ y $m$ para índices internos de sumas, y $q$ para el modo sobre el cual se proyecta con $\sin(qx)$.

## 10.2(i) — Expansión de Galerkin con simetría impar

Como en 10.1 vimos que la simetría impar se conserva, tomamos una expansión solo en senos,

$$
u(x,t)=\sum_{\ell=1}^{\infty}a_\ell(t)\sin(\ell x),
$$

en el dominio $0\leq x\leq2\pi$. El objetivo es reemplazar esta serie en $u_t=-u_{xx}-k^2u_{xxxx}-u u_x$, proyectar sobre cada modo $\sin(qx)$, y obtener una EDO para cada amplitud $a_q(t)$.

**Derivadas de la expansión.** Los coeficientes dependen solo del tiempo, de modo que

$$
u_t=\sum_{\ell=1}^{\infty}\dot a_\ell(t)\sin(\ell x),
\qquad
u_x=\sum_{\ell=1}^{\infty}\ell a_\ell(t)\cos(\ell x).
$$

Además,

$$
u_{xx}=-\sum_{\ell=1}^{\infty}\ell^2a_\ell(t)\sin(\ell x),
\qquad
u_{xxxx}=\sum_{\ell=1}^{\infty}\ell^4a_\ell(t)\sin(\ell x).
$$

**Parte lineal.** Al sustituir en $-u_{xx}-k^2u_{xxxx}$, los senos no se mezclan entre sí:

$$
-u_{xx}-k^2u_{xxxx}
=
\sum_{\ell=1}^{\infty}(\ell^2-k^2\ell^4)a_\ell(t)\sin(\ell x).
$$

Equivalentemente,

$$
\ell^2-k^2\ell^4=(1-k^2\ell^2)\ell^2.
$$

**Término no lineal.** Aquí sí aparecen índices cruzados, porque $u u_x$ es producto de dos series. Escribimos

$$
u u_x
=
\sum_{\ell=1}^{\infty}\sum_{m=1}^{\infty}
m a_\ell(t)a_m(t)\sin(\ell x)\cos(mx).
$$

Usando

$$
\sin(\ell x)\cos(mx)
=
\frac12\left[\sin((\ell+m)x)+\sin((\ell-m)x)\right],
$$

queda

$$
u u_x
=
\frac12\sum_{\ell=1}^{\infty}\sum_{m=1}^{\infty}
m a_\ell(t)a_m(t)
\left[\sin((\ell+m)x)+\sin((\ell-m)x)\right].
$$

**Proyección sobre el modo $q$.** Para obtener la ecuación de $a_q$, multiplicamos por $\sin(qx)$ e integramos en un período. Usamos

$$
\int_0^{2\pi}\sin(nx)\sin(qx)\,dx=\pi\delta_{nq}.
$$

La parte temporal da $\pi\dot a_q$, y la lineal da $\pi(q^2-k^2q^4)a_q$. Por lo tanto,

$$
\dot a_q
=
(q^2-k^2q^4)a_q
-
\frac1\pi\int_0^{2\pi}u u_x\sin(qx)\,dx.
$$

**Proyección del término no lineal.** Reemplazando la identidad trigonométrica,

$$
\int_0^{2\pi}u u_x\sin(qx)\,dx
=
\frac12\sum_{\ell=1}^{\infty}\sum_{m=1}^{\infty}
m a_\ell a_m
\int_0^{2\pi}
\left[\sin((\ell+m)x)+\sin((\ell-m)x)\right]\sin(qx)\,dx.
$$

Solo sobreviven las combinaciones cuyo argumento coincide con $q$ o $-q$.

**Aporte de $\sin((\ell+m)x)$.** Este término contribuye cuando $\ell+m=q$, es decir, $\ell=q-m$. Como $\ell\geq1$, resulta

$$
\frac12\sum_{\ell,m}m a_\ell a_m
\int_0^{2\pi}\sin((\ell+m)x)\sin(qx)\,dx
=
\frac{\pi}{2}\sum_{m=1}^{q-1}m a_{q-m}a_m.
$$

**Aporte de $\sin((\ell-m)x)$.** Hay dos casos. Si $\ell-m=q$, entonces $\ell=q+m$, y el aporte es positivo:

$$
\frac{\pi}{2}\sum_{m=1}^{\infty}m a_{q+m}a_m.
$$

Si $\ell-m=-q$, entonces $\ell=m-q$, pero $\sin(-qx)=-\sin(qx)$; por eso aparece un signo negativo:

$$
-\frac{\pi}{2}\sum_{m=q+1}^{\infty}m a_{m-q}a_m.
$$

**Forma compacta.** Los aportes con diferencia de índices se agrupan como

$$
\operatorname{sg}(q-m)a_{|q-m|}
=
\begin{cases}
+a_{q-m}, & q>m,\\
0, & q=m,\\
-a_{m-q}, & q<m.
\end{cases}
$$

Así,

$$
\frac1\pi\int_0^{2\pi}u u_x\sin(qx)\,dx
=
\frac12\sum_{m=1}^{\infty}
m a_m(t)\left[
\operatorname{sg}(q-m)a_{|q-m|}(t)+a_{q+m}(t)
\right].
$$

**Sistema modal.** Sustituyendo la proyección no lineal,

$$
\boxed{
\dot a_q(t)
=
(1-k^2q^2)q^2a_q(t)
-
\frac12\sum_{m=1}^{\infty}
m a_m(t)
\left[
\operatorname{sg}(q-m)a_{|q-m|}(t)+a_{q+m}(t)
\right]
}
$$

para $q=1,2,3,\ldots$. Esta es la ecuación de Galerkin que se usará en los incisos siguientes.

## 10.2(ii) — Linealización alrededor del equilibrio trivial

El sistema modal tiene el equilibrio trivial $a_\ell=0$ para todo $\ell$. Estudiamos su estabilidad lineal, porque esto dice qué modos empiezan a crecer cuando aumenta el tamaño del dominio.

**Punto fijo trivial.** Si $a_\ell=0$ para todo $\ell$, la parte lineal se anula y el término cuadrático también, porque contiene productos $a_m a_j$. Por lo tanto,

$$
\boxed{a_\ell^*=0\quad\forall \ell}
$$

es un punto fijo del sistema modal.

**Perturbación.** Escribimos

$$
a_\ell(t)=a_\ell^*+\varepsilon_\ell(t)=\varepsilon_\ell(t).
$$

Al reemplazar en el sistema modal,

$$
\dot\varepsilon_\ell
=
(1-k^2\ell^2)\ell^2\varepsilon_\ell
-
\frac12\sum_{m=1}^{\infty}
m\varepsilon_m
\left[
\operatorname{sg}(\ell-m)\varepsilon_{|\ell-m|}
+
\varepsilon_{\ell+m}
\right].
$$

**Linealización.** El término de la suma es cuadrático en las perturbaciones, por lo que es $O(\varepsilon^2)$. A primer orden queda

$$
\boxed{
\dot\varepsilon_\ell=(1-k^2\ell^2)\ell^2\varepsilon_\ell.
}
$$

Los modos quedan desacoplados en la aproximación lineal.

**Autovalores lineales.** Cada modo satisface

$$
\dot\varepsilon_\ell=\lambda_\ell\varepsilon_\ell,
\qquad
\lambda_\ell=(1-k^2\ell^2)\ell^2.
$$

Luego,

$$
\varepsilon_\ell(t)=\varepsilon_\ell(0)e^{(1-k^2\ell^2)\ell^2t}.
$$

**Condición de estabilidad modal.** Como $\ell^2>0$, el signo de $\lambda_\ell$ está dado por $1-k^2\ell^2$. Entonces el modo $\ell$ es estable si

$$
k\ell>1,
$$

e inestable si

$$
k\ell<1.
$$

**Condición en términos de $L$.** Como $k=2\pi/L$, la condición de inestabilidad $k\ell<1$ equivale a

$$
\frac{2\pi\ell}{L}<1
\quad\Longleftrightarrow\quad
L>2\pi\ell.
$$

En particular, el primer modo se vuelve inestable cuando

$$
\boxed{L>2\pi.}
$$

**Estabilidad del equilibrio trivial.** Para que $a_\ell=0$ sea linealmente estable, todos los modos deben ser estables. El modo más restrictivo es $\ell=1$, por lo que

$$
k>1
\quad\Longleftrightarrow\quad
L<2\pi.
$$

Por lo tanto,

$$
\boxed{L<2\pi\Rightarrow a_\ell=0\text{ es linealmente estable}},
\qquad
\boxed{L>2\pi\Rightarrow a_\ell=0\text{ es linealmente inestable}}.
$$

**8. Resumen.**

El sistema modal es

$$
\dot a_\ell(t)
=
(1-k^2\ell^2)\ell^2 a_\ell(t)
-
\frac{1}{2}
\sum_{m=1}^{\infty}
m a_m(t)
\left[
\operatorname{sg}(\ell-m)a_{|\ell-m|}(t)
+
a_{\ell+m}(t)
\right].
$$

El punto fijo trivial es

$$
a_\ell^*=0
\qquad
\forall \ell.
$$

Al linealizar alrededor de este punto fijo, el término no lineal desaparece por ser cuadrático en los coeficientes $a_m$. Queda

$$
\dot a_\ell(t)
=
(1-k^2\ell^2)\ell^2 a_\ell(t).
$$

El autovalor del modo $\ell$ es

$$
\lambda_\ell=(1-k^2\ell^2)\ell^2.
$$

Por lo tanto:

$$
\lambda_\ell<0
\iff
k\ell>1,
$$

y

$$
\lambda_\ell>0
\iff
k\ell<1.
$$

Como

$$
k=\frac{2\pi}{L},
$$

el modo $\ell$ es inestable cuando

$$
L>2\pi\ell.
$$

En particular, el primer modo pierde estabilidad cuando

$$
L>2\pi.
$$

Por lo tanto,

$$
\boxed{
L<2\pi
\Rightarrow
\text{el punto fijo trivial es linealmente estable}
}
$$

y

$$
\boxed{
L>2\pi
\Rightarrow
\text{el punto fijo trivial es linealmente inestable.}
}
$$

**Control numérico.** Como chequeo, graficamos los autovalores $\lambda_\ell=(1-k^2\ell^2)\ell^2$ para algunos valores de $L$. El cambio de signo del primer modo debe aparecer al cruzar $L=2\pi$.

```python
import numpy as np
import matplotlib.pyplot as plt

def lambdas_lineales(L, N):
    k = 2*np.pi/L
    ell = np.arange(1, N+1)
    lam = (1 - k**2 * ell**2) * ell**2
    return ell, lam

N = 12
Ls = [1.5*np.pi, 2*np.pi, 3*np.pi, 6*np.pi]

plt.figure(figsize=(8, 5))

for L in Ls:
    ell, lam = lambdas_lineales(L, N)
    plt.plot(ell, lam, marker="o", label=rf"$L={L/np.pi:.1f}\pi$")

plt.axhline(0, color="black", linewidth=1)
plt.xlabel(r"modo $\ell$")
plt.ylabel(r"$\lambda_\ell$")
plt.title(r"Autovalores lineales $\lambda_\ell=(1-k^2\ell^2)\ell^2$")
plt.legend()
plt.grid(False)
plt.show()
```

## 10.2(iii) — Truncamiento a dos modos

Ahora retenemos solo los dos primeros modos,

$$
u(x,t)\simeq a_1(t)\sin x+a_2(t)\sin(2x),
$$

lo cual equivale a tomar $a_m=0$ para $m>2$ y $a_0=0$. Esta aproximación conserva el acoplamiento no lineal mínimo entre $a_1$ y $a_2$, y será la base del análisis de puntos fijos y bifurcaciones.

**Primer modo.** Para $q=1$, la parte lineal es $(1-k^2)a_1$. En la suma no lineal solo sobreviven $m=1,2$:

$$
-\frac12 a_1 a_2
+
a_1a_2
=
\frac12a_1a_2.
$$

Así,

$$
\boxed{\dot a_1=(1-k^2)a_1+\frac12a_1a_2.}
$$

**Segundo modo.** Para $q=2$, la parte lineal es $4(1-4k^2)a_2$. El único aporte no lineal que queda dentro del truncamiento es el de $m=1$:

$$
-\frac12a_1^2.
$$

Por lo tanto,

$$
\boxed{\dot a_2=4(1-4k^2)a_2-\frac12a_1^2.}
$$

**Sistema truncado.** El modelo de dos modos es

$$
\boxed{
\begin{aligned}
\dot a_1&=(1-k^2)a_1+\frac12a_1a_2,\\
\dot a_2&=4(1-4k^2)a_2-\frac12a_1^2.
\end{aligned}
}
$$

Es invariante bajo $(a_1,a_2)\mapsto(-a_1,a_2)$, por lo que los equilibrios no triviales aparecerán en pares simétricos.

**Conclusión del truncamiento.** El primer modo puede volverse inestable al aumentar $L$, y el segundo modo aparece forzado por el término $-a_1^2/2$. A su vez, $a_2$ realimenta a $a_1$ mediante $a_1a_2/2$. Esta interacción es la que genera las ramas no triviales del inciso siguiente.

## 10.2(iv) — Puntos fijos y estabilidad del sistema truncado

Consideramos el sistema

$$
\dot a_1=(1-k^2)a_1+\frac12a_1a_2,
\qquad
\dot a_2=4(1-4k^2)a_2-\frac12a_1^2.
$$

Buscamos sus puntos fijos y los clasificamos mediante el Jacobiano. El parámetro $k=2\pi/L$ se traducirá luego a condiciones sobre el tamaño del dominio.

**Ecuaciones de punto fijo.** Imponiendo $\dot a_1=\dot a_2=0$,

$$
0=a_1\left(1-k^2+\frac{a_2}{2}\right),
\qquad
0=4(1-4k^2)a_2-\frac12a_1^2.
$$

La primera ecuación separa dos casos: $a_1=0$, o bien $a_2=2(k^2-1)$.

**Equilibrio trivial.** Si $a_1=0$, entonces $4(1-4k^2)a_2=0$. Para $k\neq1/2$ queda

$$
\boxed{P_0=(0,0).}
$$

Exactamente en $k=1/2$ aparece una degeneración: $a_1=0$ y $a_2$ arbitrario también son equilibrios. Este caso será relevante al interpretar $L=4\pi$.

**Equilibrios no triviales.** Si $a_2=2(k^2-1)$, sustituyendo en la segunda ecuación resulta

$$
a_1^2=16(k^2-1)(1-4k^2).
$$

Por lo tanto,

$$
\boxed{
P_\pm=
\left(
\pm4\sqrt{(k^2-1)(1-4k^2)},\;
2(k^2-1)
\right).
}
$$

Estos puntos son reales si $1/2\leq k\leq1$, es decir, si

$$
2\pi\leq L\leq4\pi.
$$

**Jacobiano.** Para

$$
f_1=(1-k^2)a_1+\frac12a_1a_2,
\qquad
f_2=4(1-4k^2)a_2-\frac12a_1^2,
$$

la matriz jacobiana es

$$
\boxed{
J(a_1,a_2)=
\begin{pmatrix}
(1-k^2)+a_2/2 & a_1/2\\
-a_1 & 4(1-4k^2)
\end{pmatrix}.
}
$$

**Estabilidad de $P_0$.** En el origen,

$$
J(P_0)=
\begin{pmatrix}
1-k^2 & 0\\
0 & 4(1-4k^2)
\end{pmatrix}.
$$

Entonces $\lambda_1=1-k^2$ y $\lambda_2=4(1-4k^2)$. Por lo tanto:

| Intervalo | Clasificación de $P_0$ |
|---|---|
| $k>1$ | estable |
| $1/2<k<1$ | silla |
| $0<k<1/2$ | nodo inestable |

En términos de $L$, esto corresponde a $L<2\pi$, $2\pi<L<4\pi$ y $L>4\pi$, respectivamente.

**Estabilidad de $P_\pm$.** En los puntos no triviales se cumple $a_2=2(k^2-1)$, de modo que

$$
J(P_\pm)=
\begin{pmatrix}
0 & a_1^*/2\\
-a_1^* & 4(1-4k^2)
\end{pmatrix}.
$$

La traza y el determinante son

$$
\operatorname{tr}J=4(1-4k^2),
\qquad
\det J=\frac{(a_1^*)^2}{2}=8(k^2-1)(1-4k^2).
$$

Para $1/2<k<1$, $\operatorname{tr}J<0$ y $\det J>0$, por lo que ambos puntos son estables. Más precisamente,

$$
\frac12<k<\frac1{\sqrt2}
\Rightarrow P_\pm \text{ son focos estables},
$$

mientras que

$$
\frac1{\sqrt2}<k<1
\Rightarrow P_\pm \text{ son nodos estables}.
$$

**Resumen de estabilidad.**

| Intervalo de $k$ | $P_0$ | $P_\pm$ |
|---|---|---|
| $k>1$ | estable | no existen |
| $1/\sqrt2<k<1$ | silla | nodos estables |
| $1/2<k<1/\sqrt2$ | silla | focos estables |
| $0<k<1/2$ | nodo inestable | no existen |

Los dos puntos $P_+$ y $P_-$ tienen la misma estabilidad por la simetría $(a_1,a_2)\mapsto(-a_1,a_2)$.

**Control numérico.** La siguiente celda evalúa los puntos fijos y los autovalores para algunos valores representativos de $k$. Sirve como verificación de las clasificaciones anteriores.

```python
import numpy as np


def jacobiano(a1, a2, k):
    return np.array([
        [(1-k**2) + 0.5*a2, 0.5*a1],
        [-a1, 4*(1-4*k**2)],
    ])


def puntos_fijos(k):
    puntos = [("P0", 0.0, 0.0)]
    radicando = (k**2-1)*(1-4*k**2)
    if radicando > 0:
        a1 = 4*np.sqrt(radicando)
        a2 = 2*(k**2-1)
        puntos.extend([("P+", a1, a2), ("P-", -a1, a2)])
    return puntos


for k in (1.2, 0.85, 0.60, 0.40):
    print(f"\nk = {k:.2f}")
    for nombre, a1, a2 in puntos_fijos(k):
        eig = np.linalg.eigvals(jacobiano(a1, a2, k))
        print(f"  {nombre}: ({a1:+.4f}, {a2:+.4f}), autovalores={eig}")
```

**Conclusión del inciso.** El origen existe para todo $k$. Al cruzar $k=1$, pierde estabilidad y aparecen dos equilibrios estables $P_\pm$. En $k=1/2$ el truncamiento se vuelve degenerado y aparece una línea de equilibrios con $a_1=0$.

## 10.2(v) — Diagrama de bifurcaciones

Finalmente representamos las ramas estacionarias del sistema de dos modos en función de $L$. Como $k=2\pi/L$, aumentar $L$ equivale a disminuir $k$. El diagrama pedido se hace con la coordenada $A_1^*$, pero también conviene mirar $A_2^*$ para no perder información.

**Ramas de puntos fijos.** El punto trivial es $P_0=(0,0)$ y existe para todo $L$. Los puntos no triviales son

$$
P_\pm=
\left(
\pm4\sqrt{(k^2-1)(1-4k^2)},
\;2(k^2-1)
\right),
\qquad
k=\frac{2\pi}{L}.
$$

Por lo tanto,

$$
A_{1,\pm}^*(L)=
\pm4\sqrt{
\left[\left(\frac{2\pi}{L}\right)^2-1\right]
\left[1-4\left(\frac{2\pi}{L}\right)^2\right]
},
$$

y existen solo para $2\pi\leq L\leq4\pi$.

**Estabilidad sobre el diagrama.** La clasificación en términos de $L$ es:

| Intervalo de $L$ | $P_0$ | $P_\pm$ |
|---|---|---|
| $L<2\pi$ | estable | no existen |
| $2\pi<L<2\pi\sqrt2$ | silla | nodos estables |
| $2\pi\sqrt2<L<4\pi$ | silla | focos estables |
| $L>4\pi$ | nodo inestable | no existen |

El valor $L=2\pi\sqrt2$ solo cambia la forma de convergencia hacia $P_\pm$: de nodo a foco.

```python
import numpy as np
import matplotlib.pyplot as plt


def ramas_no_triviales(L):
    """Devuelve A1+, A1- y A2 de P±; fuera de existencia devuelve NaN."""
    L = np.asarray(L, dtype=float)
    k = 2*np.pi/L
    radicando = (k**2-1)*(1-4*k**2)
    existe = (L >= 2*np.pi) & (L <= 4*np.pi)

    A1p = np.full_like(L, np.nan)
    A2 = np.full_like(L, np.nan)
    A1p[existe] = 4*np.sqrt(np.maximum(radicando[existe], 0))
    A2[existe] = 2*(k[existe]**2-1)
    return A1p, -A1p, A2


L = np.linspace(1.0, 14.0, 1800)
A1p, A1m, A2nt = ramas_no_triviales(L)

Lc1 = 2*np.pi
L_nodo_foco = 2*np.pi*np.sqrt(2)
Lc2 = 4*np.pi
```

**Diagrama de $A_1^*$.** En el gráfico se marca el origen con distintos estilos según su estabilidad y las ramas no triviales con colores distintos según sean nodos o focos estables. Las dos ramas son simétricas porque el sistema no cambia bajo $a_1\mapsto -a_1$.

```python
fig, ax = plt.subplots(figsize=(10.5, 6.2), constrained_layout=True)

origen_estable = L < Lc1
origen_silla = (L >= Lc1) & (L < Lc2)
origen_inestable = L >= Lc2
nodos = (L >= Lc1) & (L < L_nodo_foco)
focos = (L >= L_nodo_foco) & (L <= Lc2)

ax.plot(L[origen_estable], np.zeros(origen_estable.sum()),
        color="#1565c0", lw=3, label=r"$P_0$: estable")
ax.plot(L[origen_silla], np.zeros(origen_silla.sum()),
        color="#ef6c00", ls="--", lw=2.5, label=r"$P_0$: silla")
ax.plot(L[origen_inestable], np.zeros(origen_inestable.sum()),
        color="#c62828", ls=":", lw=3, label=r"$P_0$: nodo inestable")

ax.plot(L[nodos], A1p[nodos], color="#2e7d32", lw=3,
        label=r"$P_\pm$: nodos estables")
ax.plot(L[nodos], A1m[nodos], color="#2e7d32", lw=3)
ax.plot(L[focos], A1p[focos], color="#7b1fa2", lw=3,
        label=r"$P_\pm$: focos estables")
ax.plot(L[focos], A1m[focos], color="#7b1fa2", lw=3)

for xcrit, texto in ((Lc1, r"$2\pi$"),
                     (L_nodo_foco, r"$2\pi\sqrt{2}$"),
                     (Lc2, r"$4\pi$")):
    ax.axvline(xcrit, color="0.35", ls=":", lw=1.3)
    ax.text(xcrit+0.06, 4.05, texto, rotation=90, va="top", fontsize=10)

ax.set(xlim=(1, 14), ylim=(-4.5, 4.5),
       xlabel=r"tamaño del dominio $L$", ylabel=r"amplitud $A_1^*$",
       title=r"Diagrama de bifurcaciones del truncamiento a dos modos")
ax.grid(alpha=0.22)
ax.legend(frameon=False, fontsize=9, ncol=2, loc="lower left")
plt.show()
```

**Primera transición.** En $L=2\pi$, los puntos $P_\pm$ nacen desde el origen y $P_0$ cambia de estable a silla. Esta es una bifurcación pitchfork supercrítica. Un parámetro efectivo es

$$
\mu=1-k^2=1-\left(\frac{2\pi}{L}\right)^2,
$$

que cambia de signo en $L=2\pi$.

**Qué ocurre en $L=4\pi$.** En el gráfico de $A_1^*$, las ramas no triviales vuelven a $A_1^*=0$. Sin embargo, en el plano completo terminan en

$$
P_\pm\longrightarrow\left(0,-\frac32\right),
$$

mientras que $P_0=(0,0)$. Por eso no chocan nuevamente con el origen. En $L=4\pi$, equivalente a $k=1/2$, aparece una línea degenerada de equilibrios con $a_1=0$.

**Segunda coordenada.** El siguiente gráfico no es estrictamente pedido, pero evita una interpretación equivocada del extremo $L=4\pi$: las ramas no triviales parten del origen en $L=2\pi$, pero terminan en $A_2^*=-3/2$.

```python
fig, ax = plt.subplots(figsize=(10.5, 5.5), constrained_layout=True)

ax.plot(L, np.zeros_like(L), color="#1565c0", lw=2.5,
        label=r"$P_0$: $A_2^*=0$")
ax.plot(L, A2nt, color="#6a1b9a", lw=3,
        label=r"$P_\pm$: $A_2^*=2[(2\pi/L)^2-1]$")

ax.scatter([Lc1, Lc2], [0, -1.5], color="black", zorder=4)
ax.annotate(r"$P_\pm\to P_0$", (Lc1, 0), xytext=(Lc1+0.5, -0.4),
            arrowprops={"arrowstyle": "->"})
ax.annotate(r"$P_\pm\to(0,-3/2)$", (Lc2, -1.5),
            xytext=(Lc2-3.0, -1.1), arrowprops={"arrowstyle": "->"})
ax.axvline(Lc1, color="0.4", ls=":")
ax.axvline(Lc2, color="0.4", ls=":")

ax.set(xlim=(1, 14), ylim=(-1.8, 0.35),
       xlabel=r"tamaño del dominio $L$", ylabel=r"segunda amplitud $A_2^*$",
       title=r"Segunda coordenada de las ramas de equilibrio")
ax.grid(alpha=0.22)
ax.legend(frameon=False)
plt.show()
```

**Verificación con autovalores.** Evaluamos el Jacobiano en algunos valores de $L$ para confirmar numéricamente los cambios de estabilidad y la transición nodo–foco.

```python
def jacobiano(A1, A2, L):
    k = 2*np.pi/L
    return np.array([
        [(1-k**2)+A2/2, A1/2],
        [-A1, 4*(1-4*k**2)],
    ])


for L_prueba in (1.8*np.pi, 2.2*np.pi, 3*np.pi, 4.2*np.pi):
    A1p_, A1m_, A2_ = ramas_no_triviales(np.array([L_prueba]))
    eig0 = np.linalg.eigvals(jacobiano(0, 0, L_prueba))
    print(f"L/pi={L_prueba/np.pi:.2f}: P0 -> {eig0}")
    if np.isfinite(A1p_[0]):
        eigp = np.linalg.eigvals(jacobiano(A1p_[0], A2_[0], L_prueba))
        print(f"             P± -> {eigp}")
```

**Conclusión de 10.2.** El truncamiento a dos modos captura la primera pérdida de estabilidad del estado trivial. En $L=2\pi$ aparece una pitchfork supercrítica: el origen deja de ser estable y nacen dos ramas simétricas estables. En $L=4\pi$ aparece una degeneración propia del truncamiento, no una segunda pitchfork ordinaria. A medida que $L$ sigue aumentando, entran en juego más modos y el modelo de dos amplitudes deja de ser suficiente, lo que motiva las simulaciones numéricas de 10.3.

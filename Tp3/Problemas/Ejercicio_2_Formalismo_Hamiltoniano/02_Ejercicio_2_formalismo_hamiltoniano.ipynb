### **Ejercicio 2: Formalismo Hamiltoniano**

En base al análisis anterior, vamos a plantear una Hamiltoniano de interacción a 4 ondas dado por
 $$ \mathcal{H} = \int \omega_k |a_k|^2~\text{d}k + \int g ~a_k^* a_1^* a_2 a_3 ~\delta(k + k_1 - k_2 - k_3)~\text{d}k_1\text{d}k_2\text{d}k_3\text{d}k $$
donde $g$ es una amplitud de _scattering_ constante, $a$ es la amplitud compleja asociada y $w_k$ la relación de dispersión del ejercicio anterior.
Recuerde que puede deshacerse del término lineal en $k$ de $\omega_k$ mediante una transformación de Galileo, por lo que será $\omega_k\propto k^2$.

(a) Usando las ecuaciones de Hamilton, encuentre la ecuación diferencial para $\dot{a}_k$.

(b) Muestre que la ecuación diferencial en derivadas parciales asociada a este Hamiltoniano es
$$i\frac{∂a}{∂t} = -α\nabla^2a + g|a|²a$$
para un $α>0$ dependiente de $ω_k$.
¿Puede esta ecuación representar un campo escalar real?

(c) Muestre que $E = \int |a(x,t)|^2 \text{d} x$ es una magnitud conservada. ¿Que tipos de dominios y/o condiciones de contorno la vuelven constante?

(d) Plantee una solución tipo onda viajera $a(x,t) = \phi(x-vt) \exp(i\xi)$ para $\xi = \kappa x - \Omega t$ y encuentre los solitones del problema. Considere por separado los casos $g<0$ (solitones brillantes) y $g>0$ (solitones oscuros). Grafique los $\phi(\xi)$ hallados. ¿Por qué se los llama así?
Muestre que un $\phi$ constante también es solución, por lo que una onda monocromática se mantiene monocromática.
 _Ayuda: Una ecuación compleja equivale a dos reales, por lo que además podrá determinar el valor de $v$ y $\Omega$ en función de $\kappa$ y la amplitud de $\phi$_.

## **Parte numérica**

Para pasar a la parte numérica, vamos a considerar un dominio periódico de longitud $2\pi L$, tal que el mínimo vector de onda es $k_\text{min} = 1/L$. Esto nos permite inmediatamente definir análogos adimensionales $x' = x/L$ y $k' = kL$ tales que $0 \leq x < 2\pi$ y $k'$ sea entero.
En particular, trabajaremos con $N=128$ puntos de grilla, por lo que
$$k'\in\{-N/2, -N/2+1,.., -1, 0, 1, .., N/2-2, N/2-1\}.$$
Dada la no linealidad del problema, el máximo $k$ que podremos resolver es $k_\text{max} = N/4 = 32$.
Además, en este caso discreto, la energía introducida en el ejercicio 2(c) toma la forma
$$ E(t) = \frac{1}{N}\sum_{j=0}^{N-1} |a(x_j, t)|^2 $$
donde $x_j = 2\pi j/N$ son los puntos de grilla.
En ```Python```, puede computarse usando ```numpy.mean``` sobre el módulo cuadrado del arreglo real.
Analizar la conservación de esta energía en las simulaciones es un buen diagnóstico para verificar la correctitud del código.

# Problema 10.1 — Propiedades generales de Kuramoto-Sivashinsky

Consideramos la ecuación de Kuramoto-Sivashinsky en una dimensión:

$$
u_t=-u_{xx}-u_{xxxx}-u u_x,
$$

con condiciones periódicas en un dominio

$$
0\leq x\leq L.
$$

También puede escribirse como

$$
u_t+u u_x+u_{xx}+u_{xxxx}=0.
$$

En este notebook se resuelven los incisos del punto 10.1:

1. Conservación del promedio espacial.
2. Invariancia galileana.
3. Simetría de reflexión.
4. Reescalamiento del dominio \(0\leq x'\leq L\) al dominio \(0\leq x\leq 2\pi\).

# 10.1(i) Conservación del promedio espacial

Definimos

$$
M(t)=\int_0^L u(x,t)\,dx.
$$

Derivando respecto del tiempo,

$$
\frac{dM}{dt}
=
\int_0^L u_t(x,t)\,dx.
$$

Usando la ecuación

$$
u_t=-u_{xx}-u_{xxxx}-u u_x,
$$

obtenemos

$$
\frac{dM}{dt}
=
-\int_0^L u_{xx}\,dx
-\int_0^L u_{xxxx}\,dx
-\int_0^L u u_x\,dx.
$$

El primer término es

$$
-\int_0^L u_{xx}\,dx
=
-\left[u_x\right]_0^L=0,
$$

porque \(u_x\) es periódica.

El segundo término es

$$
-\int_0^L u_{xxxx}\,dx
=
-\left[u_{xxx}\right]_0^L=0,
$$

porque \(u_{xxx}\) también es periódica.

Para el término no lineal usamos

$$
u u_x=\frac{\partial}{\partial x}\left(\frac{u^2}{2}\right).
$$

Entonces

$$
-\int_0^L u u_x\,dx
=
-\left[\frac{u^2}{2}\right]_0^L=0,
$$

porque \(u\) es periódica.

Por lo tanto,

$$
\boxed{
\frac{dM}{dt}=0.
}
$$

Así,

$$
\boxed{
M(t)=\int_0^L u(x,t)\,dx
}
$$

es constante en el tiempo. En particular, el promedio espacial

$$
\overline u(t)=\frac{1}{L}\int_0^L u(x,t)\,dx
$$

también se conserva.

# 10.1(ii) Invariancia galileana

Tomamos una constante \(U\) y definimos

$$
x'=x+Ut,
\qquad
t'=t,
\qquad
u'(x',t')=u(x,t)+U.
$$

Como

$$
x=x'-Ut',
\qquad
t=t',
$$

podemos escribir

$$
u'(x',t')=u(x'-Ut',t')+U.
$$

Queremos ver que si

$$
u_t+u u_x+u_{xx}+u_{xxxx}=0,
$$

entonces

$$
u'_{t'}+u'u'_{x'}+u'_{x'x'}+u'_{x'x'x'x'}=0.
$$

## Derivadas bajo la transformación galileana

Para la derivada espacial,

$$
u'_{x'}
=
\frac{\partial}{\partial x'}
\left[
u(x'-Ut',t')+U
\right].
$$

Por regla de la cadena,

$$
u'_{x'}
=
u_x\frac{\partial (x'-Ut')}{\partial x'}
+
u_t\frac{\partial t'}{\partial x'}.
$$

Como

$$
\frac{\partial (x'-Ut')}{\partial x'}=1,
\qquad
\frac{\partial t'}{\partial x'}=0,
$$

queda

$$
\boxed{
u'_{x'}=u_x.
}
$$

Por lo tanto,

$$
\boxed{
u'_{x'x'}=u_{xx},
\qquad
u'_{x'x'x'x'}=u_{xxxx}.
}
$$

Para la derivada temporal,

$$
u'_{t'}
=
\frac{\partial}{\partial t'}
\left[
u(x'-Ut',t')+U
\right].
$$

Por regla de la cadena,

$$
u'_{t'}
=
u_x\frac{\partial (x'-Ut')}{\partial t'}
+
u_t\frac{\partial t'}{\partial t'}.
$$

Como

$$
\frac{\partial (x'-Ut')}{\partial t'}=-U,
\qquad
\frac{\partial t'}{\partial t'}=1,
$$

tenemos

$$
\boxed{
u'_{t'}=u_t-Uu_x.
}
$$

## Sustitución en la ecuación

Sustituyendo en la ecuación primada,

$$
u'_{t'}+u'u'_{x'}+u'_{x'x'}+u'_{x'x'x'x'}
=
(u_t-Uu_x)+(u+U)u_x+u_{xx}+u_{xxxx}.
$$

Distribuyendo,

$$
=
u_t-Uu_x+u u_x+Uu_x+u_{xx}+u_{xxxx}.
$$

Los términos galileanos se cancelan:

$$
-Uu_x+Uu_x=0.
$$

Entonces

$$
u'_{t'}+u'u'_{x'}+u'_{x'x'}+u'_{x'x'x'x'}
=
u_t+u u_x+u_{xx}+u_{xxxx}.
$$

Si \(u\) satisface la ecuación original, el lado derecho es cero. Luego,

$$
\boxed{
u'_{t'}+u'u'_{x'}+u'_{x'x'}+u'_{x'x'x'x'}=0.
}
$$

Por lo tanto, la ecuación es invariante ante transformaciones galileanas.

## Comentario sobre el promedio nulo

Como el promedio espacial se conserva y la transformación galileana cambia

$$
u\mapsto u+U,
$$

podemos elegir

$$
U=-\overline u
$$

para trabajar en un sistema donde el promedio espacial sea cero. Por eso puede suponerse, sin pérdida de generalidad, que

$$
\int_0^L u(x,t)\,dx=0.
$$

# 10.1(iii) Simetría de reflexión

Consideramos la transformación

$$
u'(x,t)=-u(-x,t).
$$

Queremos probar que si \(u\) satisface

$$
u_t=-u_{xx}-u_{xxxx}-u u_x,
$$

entonces \(u'\) también satisface la misma ecuación.

Calculamos las derivadas:

$$
u'_t(x,t)=-u_t(-x,t),
$$

$$
u'_x(x,t)=u_x(-x,t),
$$

$$
u'_{xx}(x,t)=-u_{xx}(-x,t),
$$

y

$$
u'_{xxxx}(x,t)=-u_{xxxx}(-x,t).
$$

Además,

$$
u'(x,t)u'_x(x,t)
=
[-u(-x,t)][u_x(-x,t)]
=
-u(-x,t)u_x(-x,t).
$$

Ahora calculamos el lado derecho de la ecuación para \(u'\):

$$
-u'_{xx}-u'_{xxxx}-u'u'_x.
$$

Reemplazando,

$$
-u'_{xx}-u'_{xxxx}-u'u'_x
=
u_{xx}(-x,t)+u_{xxxx}(-x,t)+u(-x,t)u_x(-x,t).
$$

Como \(u\) satisface la ecuación original,

$$
u_t(-x,t)
=
-u_{xx}(-x,t)-u_{xxxx}(-x,t)-u(-x,t)u_x(-x,t).
$$

Multiplicando por \(-1\),

$$
-u_t(-x,t)
=
u_{xx}(-x,t)+u_{xxxx}(-x,t)+u(-x,t)u_x(-x,t).
$$

Pero

$$
u'_t(x,t)=-u_t(-x,t).
$$

Entonces

$$
u'_t=-u'_{xx}-u'_{xxxx}-u'u'_x.
$$

Por lo tanto,

$$
\boxed{
u'(x,t)=-u(-x,t)
}
$$

también es solución.

## Consecuencia: subespacio impar

Si la condición inicial es impar,

$$
u(-x,0)=-u(x,0),
$$

entonces la evolución conserva esa paridad.

En ese caso se puede trabajar con una expansión en senos:

$$
u(x,t)=\sum_{\ell=1}^{\infty}a_\ell(t)\sin(\ell x).
$$

Esto conecta directamente con el inciso 10.2.

# 10.1(iv) Reescalamiento del dominio

Ahora partimos de la ecuación en variables primadas:

$$
u'_{t'}=-u'_{x'x'}-u'_{x'x'x'x'}-u'u'_{x'}.
$$

El dominio original es

$$
0\leq x'\leq L.
$$

Queremos llevarlo al dominio normalizado

$$
0\leq x\leq 2\pi.
$$

Definimos

$$
k=\frac{2\pi}{L}
$$

y tomamos

$$
x=kx'.
$$

Entonces

$$
x'=0\Rightarrow x=0,
$$

y

$$
x'=L\Rightarrow x=kL=2\pi.
$$

Además introducimos una escala temporal general

$$
t=\alpha t',
$$

y una escala de amplitud

$$
u'=ku.
$$

Entonces

$$
u'(x',t')=k\,u(kx',\alpha t').
$$

La idea es aplicar regla de la cadena primero y fijar \(\alpha\) después.

## Derivadas por regla de la cadena

Para el tiempo,

$$
u'_{t'}
=
\frac{\partial}{\partial t'}
\left[
k\,u(kx',\alpha t')
\right].
$$

Como \(k\) es constante,

$$
u'_{t'}=k u_t\frac{\partial t}{\partial t'}.
$$

Dado que

$$
t=\alpha t',
$$

obtenemos

$$
\frac{\partial t}{\partial t'}=\alpha.
$$

Por lo tanto,

$$
\boxed{
u'_{t'}=k\alpha u_t.
}
$$

Para el espacio,

$$
u'_{x'}
=
\frac{\partial}{\partial x'}
\left[
k\,u(kx',\alpha t')
\right].
$$

Por regla de la cadena,

$$
u'_{x'}=k u_x\frac{\partial x}{\partial x'}.
$$

Como

$$
x=kx',
$$

tenemos

$$
\frac{\partial x}{\partial x'}=k.
$$

Entonces

$$
\boxed{
u'_{x'}=k^2u_x.
}
$$

Cada derivada espacial adicional introduce otro factor \(k\). Por lo tanto,

$$
\boxed{
u'_{x'x'}=k^3u_{xx},
}
$$

y

$$
\boxed{
u'_{x'x'x'x'}=k^5u_{xxxx}.
}
$$

Finalmente,

$$
u'u'_{x'}=(ku)(k^2u_x).
$$

Así,

$$
\boxed{
u'u'_{x'}=k^3u u_x.
}
$$

## Sustitución en la ecuación

Reemplazamos en

$$
u'_{t'}=-u'_{x'x'}-u'_{x'x'x'x'}-u'u'_{x'}.
$$

Usando los resultados anteriores,

$$
k\alpha u_t
=
-k^3u_{xx}
-k^5u_{xxxx}
-k^3u u_x.
$$

Dividiendo por \(k\alpha\),

$$
u_t
=
-\frac{k^2}{\alpha}u_{xx}
-
\frac{k^4}{\alpha}u_{xxxx}
-
\frac{k^2}{\alpha}u u_x.
$$

Hasta aquí, \(\alpha\) es arbitrario.

## Elección de la escala temporal

Elegimos \(\alpha\) para dejar el coeficiente de \(u_{xx}\) igual a \(-1\). Pedimos

$$
\frac{k^2}{\alpha}=1.
$$

Entonces

$$
\alpha=k^2.
$$

Por lo tanto,

$$
t=k^2t'.
$$

Con esta elección,

$$
u_t=-u_{xx}-k^2u_{xxxx}-u u_x.
$$

Así, la ecuación reescalada queda

$$
\boxed{
u_t=-u_{xx}-k^2u_{xxxx}-u u_x,
}
$$

con

$$
\boxed{
k=\frac{2\pi}{L}.
}
$$

## Interpretación

El reescalamiento

$$
x=kx',
\qquad
t=k^2t',
\qquad
u'=ku,
\qquad
k=\frac{2\pi}{L},
$$

lleva el dominio \(0\leq x'\leq L\) al dominio \(0\leq x\leq 2\pi\).

El tamaño del dominio queda incorporado en el coeficiente del término de cuarto orden:

$$
-k^2u_{xxxx}.
$$

Como

$$
k=\frac{2\pi}{L},
$$

aumentar \(L\) disminuye \(k\), y por lo tanto reduce el peso relativo del término estabilizante de cuarto orden.

# Resumen final

Para la ecuación

$$
u_t=-u_{xx}-u_{xxxx}-u u_x
$$

con condiciones periódicas:

1. Se conserva la integral espacial:

$$
\boxed{
\frac{d}{dt}\int_0^L u(x,t)\,dx=0.
}
$$

2. Es invariante bajo transformaciones galileanas:

$$
x'=x+Ut,
\qquad
t'=t,
\qquad
u'(x',t')=u(x,t)+U.
$$

3. Es invariante bajo la reflexión

$$
u'(x,t)=-u(-x,t).
$$

4. Bajo el reescalamiento

$$
k=\frac{2\pi}{L},
\qquad
x=kx',
\qquad
t=k^2t',
\qquad
u'=ku,
$$

la ecuación queda

$$
\boxed{
u_t=-u_{xx}-k^2u_{xxxx}-u u_x.
}
$$

```python
# Verificación visual del cambio de dominio x = k x'
# Si x' está en [0,L], entonces x está en [0,2*pi].

import numpy as np
import matplotlib.pyplot as plt

L = 10.0
k = 2*np.pi/L

xp = np.linspace(0, L, 300)
x = k*xp

plt.figure(figsize=(7, 4))
plt.plot(xp, x)
plt.axhline(2*np.pi, linestyle="--", linewidth=1)
plt.axvline(L, linestyle="--", linewidth=1)
plt.xlabel(r"$x'$")
plt.ylabel(r"$x=kx'$")
plt.title(r"Reescalamiento: $[0,L]\to[0,2\pi]$")
plt.grid(False)
plt.show()
```

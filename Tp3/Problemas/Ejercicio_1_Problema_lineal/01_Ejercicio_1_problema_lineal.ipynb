# Ejercicio 1: Estudio del problema lineal

Consideramos dos fluidos incompresibles, ideales e irrotacionales separados por una interfaz. Tomamos el eje vertical $y$ de modo que la interfaz de equilibrio esté en

$$
y=0,
$$

con el fluido 1 en la región superior,

$$
0<y<h_1,
$$

y el fluido 2 en la región inferior,

$$
-h_2<y<0.
$$

La perturbación de la interfaz se escribe como

$$
y=\eta(x,t).
$$

Como el flujo es irrotacional, introducimos potenciales de velocidad en cada región:

$$
\mathbf v_1=\nabla \phi_1,
\qquad
\mathbf v_2=\nabla \phi_2.
$$

Además, como los fluidos son incompresibles,

$$
\nabla\cdot \mathbf v_j=0,
$$

y por lo tanto cada potencial satisface la ecuación de Laplace:

$$
\nabla^2\phi_j=0,
\qquad j=1,2.
$$

## (a) Condiciones de contorno para una perturbación armónica

La interfaz se mueve junto con el fluido. Por lo tanto, la condición cinemática en la interfaz es que la velocidad vertical de cada fluido coincida con la velocidad vertical de la interfaz.

Linealizando alrededor de $y=0$, queda

$$
v_{1y}(x,0,t)=v_{2y}(x,0,t)=\frac{\partial \eta}{\partial t}(x,t).
$$

Como $v_{jy}=\partial_y\phi_j$, esto equivale a

$$
\frac{\partial \phi_1}{\partial y}(x,0,t)
=
\frac{\partial \phi_2}{\partial y}(x,0,t)
=
\frac{\partial \eta}{\partial t}(x,t).
$$

En los bordes rígidos superior e inferior no puede haber flujo normal. Por lo tanto,

$$
\frac{\partial \phi_1}{\partial y}(x,h_1,t)=0,
$$

$$
\frac{\partial \phi_2}{\partial y}(x,-h_2,t)=0.
$$

Para escribir la condición dinámica usamos Bernoulli no estacionario. Para cada fluido,

$$
\frac{\partial \phi_j}{\partial t}
+\frac{p_j}{\rho_j}
+gy
+\frac{|\nabla \phi_j|^2}{2}
=C_j(t).
$$

Linealizamos alrededor del equilibrio. El término cuadrático $|\nabla\phi_j|^2/2$ es de segundo orden y se descarta:

$$
\frac{\partial \phi_j}{\partial t}
+\frac{p_j}{\rho_j}
+gy
=C_j(t).
$$

Entonces

$$
p_j
=
-\rho_j\frac{\partial \phi_j}{\partial t}
-\rho_j gy
+\rho_j C_j(t).
$$

Evaluamos en la interfaz $y=\eta(x,t)$. A primer orden, podemos evaluar los potenciales en $y=0$, pero debemos conservar el término hidrostático proporcional a $\eta$:

$$
p_j(x,\eta,t)
=
-\rho_j\frac{\partial \phi_j}{\partial t}(x,0,t)
-\rho_j g\eta(x,t)
+\rho_j C_j(t).
$$

La presión es continua a través de la interfaz:

$$
p_1(x,\eta,t)=p_2(x,\eta,t).
$$

Por lo tanto,

$$
\rho_1\frac{\partial \phi_1}{\partial t}(x,0,t)
-
\rho_2\frac{\partial \phi_2}{\partial t}(x,0,t)
+
(\rho_1-\rho_2)g\eta(x,t)
+
\rho_2 C_2(t)-\rho_1 C_1(t)
=0.
$$

Las funciones $C_j(t)$ no modifican el campo de velocidades, de modo que pueden absorberse en la definición del potencial. Tomando esa libertad de gauge,

$$
\rho_2 C_2(t)-\rho_1 C_1(t)=0.
$$

Así, la condición dinámica linealizada queda

$$
\rho_1\frac{\partial \phi_1}{\partial t}(x,0,t)
-
\rho_2\frac{\partial \phi_2}{\partial t}(x,0,t)
+
(\rho_1-\rho_2)g\eta(x,t)
=0.
$$

Ahora proponemos una perturbación armónica con número de onda horizontal $k$:

$$
\eta(x,t)=\eta_0 e^{i(kx-\omega t)},
$$

$$
\phi_1(x,y,t)=\Phi_1(y)e^{i(kx-\omega t)},
$$

$$
\phi_2(x,y,t)=\Phi_2(y)e^{i(kx-\omega t)}.
$$

Con esta forma,

$$
\frac{\partial \eta}{\partial t}
=
-i\omega \eta_0 e^{i(kx-\omega t)}.
$$

Entonces las condiciones cinemáticas en la interfaz se transforman en

$$
\frac{d\Phi_1}{dy}(0)
=
\frac{d\Phi_2}{dy}(0)
=
-i\omega \eta_0.
$$

La condición dinámica queda

$$
-i\omega \rho_1\Phi_1(0)
+i\omega \rho_2\Phi_2(0)
+
(\rho_1-\rho_2)g\eta_0
=0.
$$

Finalmente, las condiciones de borde rígido son

$$
\frac{d\Phi_1}{dy}(h_1)=0,
$$

$$
\frac{d\Phi_2}{dy}(-h_2)=0.
$$

Por lo tanto, las cuatro condiciones de contorno pedidas son

$$
\boxed{
\frac{d\Phi_1}{dy}(0)
=
\frac{d\Phi_2}{dy}(0)
=
-i\omega \eta_0
}
$$

$$
\boxed{
-i\omega \rho_1\Phi_1(0)
+i\omega \rho_2\Phi_2(0)
+
(\rho_1-\rho_2)g\eta_0
=0
}
$$

$$
\boxed{
\frac{d\Phi_1}{dy}(h_1)=0
}
$$

$$
\boxed{
\frac{d\Phi_2}{dy}(-h_2)=0
}
$$

## (b) Relación de dispersión

Como cada potencial satisface la ecuación de Laplace,

$$
\nabla^2\phi_j=0.
$$

Usando

$$
\phi_j(x,y,t)=\Phi_j(y)e^{i(kx-\omega t)},
$$

obtenemos

$$
\frac{\partial^2 \phi_j}{\partial x^2}
+
\frac{\partial^2 \phi_j}{\partial y^2}
=0.
$$

Como

$$
\frac{\partial^2 \phi_j}{\partial x^2}
=
-k^2\Phi_j(y)e^{i(kx-\omega t)},
$$

queda

$$
\left(
\frac{d^2\Phi_j}{dy^2}
-k^2\Phi_j
\right)e^{i(kx-\omega t)}
=0.
$$

Por lo tanto,

$$
\frac{d^2\Phi_j}{dy^2}
-k^2\Phi_j
=0.
$$

Resolvemos esta ecuación usando exponenciales:

$$
\Phi_j(y)=A_j e^{ky}+B_j e^{-ky}.
$$

### Potencial en el fluido superior

Para el fluido superior,

$$
\Phi_1(y)=A_1e^{ky}+B_1e^{-ky}.
$$

La condición de borde en $y=h_1$ es

$$
\frac{d\Phi_1}{dy}(h_1)=0.
$$

Como

$$
\frac{d\Phi_1}{dy}
=
kA_1e^{ky}-kB_1e^{-ky},
$$

entonces

$$
kA_1e^{kh_1}-kB_1e^{-kh_1}=0.
$$

De aquí,

$$
B_1=A_1e^{2kh_1}.
$$

Entonces

$$
\Phi_1(y)
=
A_1
\left(
e^{ky}+e^{2kh_1-ky}
\right).
$$

Usamos ahora la condición cinemática en $y=0$:

$$
\frac{d\Phi_1}{dy}(0)=-i\omega \eta_0.
$$

Pero

$$
\frac{d\Phi_1}{dy}(0)
=
kA_1(1-e^{2kh_1}).
$$

Luego,

$$
kA_1(1-e^{2kh_1})
=
-i\omega \eta_0,
$$

y por lo tanto

$$
A_1
=
\frac{-i\omega \eta_0}
{k(1-e^{2kh_1})}.
$$

Así,

$$
\Phi_1(0)
=
\frac{-i\omega \eta_0}
{k(1-e^{2kh_1})}
\left(
1+e^{2kh_1}
\right).
$$

Usando

$$
\frac{1+e^{2z}}{1-e^{2z}}
=
-\coth z,
$$

queda

$$
\Phi_1(0)
=
\frac{i\omega \eta_0}{k}
\coth(kh_1).
$$

### Potencial en el fluido inferior

Para el fluido inferior,

$$
\Phi_2(y)=A_2e^{ky}+B_2e^{-ky}.
$$

La condición de borde inferior es

$$
\frac{d\Phi_2}{dy}(-h_2)=0.
$$

Entonces

$$
kA_2e^{-kh_2}-kB_2e^{kh_2}=0.
$$

De aquí,

$$
B_2=A_2e^{-2kh_2}.
$$

Por lo tanto,

$$
\Phi_2(y)
=
A_2
\left(
e^{ky}+e^{-2kh_2-ky}
\right).
$$

Usamos la condición cinemática en $y=0$:

$$
\frac{d\Phi_2}{dy}(0)=-i\omega\eta_0.
$$

Como

$$
\frac{d\Phi_2}{dy}(0)
=
kA_2(1-e^{-2kh_2}),
$$

resulta

$$
A_2
=
\frac{-i\omega \eta_0}
{k(1-e^{-2kh_2})}.
$$

Entonces

$$
\Phi_2(0)
=
\frac{-i\omega \eta_0}
{k(1-e^{-2kh_2})}
\left(
1+e^{-2kh_2}
\right).
$$

Usando

$$
\frac{1+e^{-2z}}{1-e^{-2z}}
=
\coth z,
$$

obtenemos

$$
\Phi_2(0)
=
-\frac{i\omega\eta_0}{k}
\coth(kh_2).
$$

Sustituimos ahora $\Phi_1(0)$ y $\Phi_2(0)$ en la condición dinámica:

$$
-i\omega \rho_1\Phi_1(0)
+i\omega \rho_2\Phi_2(0)
+
(\rho_1-\rho_2)g\eta_0
=0.
$$

Usamos

$$
\Phi_1(0)
=
\frac{i\omega \eta_0}{k}
\coth(kh_1),
$$

$$
\Phi_2(0)
=
-\frac{i\omega \eta_0}{k}
\coth(kh_2).
$$

Entonces

$$
-i\omega\rho_1
\left[
\frac{i\omega\eta_0}{k}\coth(kh_1)
\right]
+
i\omega\rho_2
\left[
-\frac{i\omega\eta_0}{k}\coth(kh_2)
\right]
+
(\rho_1-\rho_2)g\eta_0
=0.
$$

Como $-i\,i=1$ e $i(-i)=1$, queda

$$
\rho_1\frac{\omega^2\eta_0}{k}\coth(kh_1)
+
\rho_2\frac{\omega^2\eta_0}{k}\coth(kh_2)
+
(\rho_1-\rho_2)g\eta_0
=0.
$$

Dividiendo por $\eta_0$,

$$
\frac{\omega^2}{k}
\left[
\rho_1\coth(kh_1)
+
\rho_2\coth(kh_2)
\right]
+
(\rho_1-\rho_2)g
=0.
$$

Por lo tanto,

$$
\omega^2
=
\frac{(\rho_2-\rho_1)gk}
{\rho_1\coth(kh_1)+\rho_2\coth(kh_2)}.
$$

La relación de dispersión es entonces

$$
\boxed{
\omega_k
=
\sqrt{
\frac{(\rho_2-\rho_1)gk}
{\rho_1\coth(kh_1)+\rho_2\coth(kh_2)}
}
}
$$

para el caso estable $\rho_2>\rho_1$.

### Límites conocidos

**1. Dos fluidos profundos.** Si

$$
kh_1\to\infty,
\qquad
kh_2\to\infty,
$$

entonces

$$
\coth(kh_1)\to 1,
\qquad
\coth(kh_2)\to 1.
$$

Así,

$$
\omega^2
=
\frac{(\rho_2-\rho_1)gk}{\rho_1+\rho_2}.
$$

Este es el resultado usual para ondas de gravedad interfaciales entre dos fluidos profundos.

**2. Fluido superior de densidad despreciable.** Si $\rho_1=0$,

$$
\omega^2
=
\frac{\rho_2 gk}{\rho_2\coth(kh_2)}
=
gk\tanh(kh_2).
$$

Si además el fluido inferior es profundo,

$$
kh_2\to\infty,
\qquad
\tanh(kh_2)\to 1,
$$

y recuperamos

$$
\omega^2=gk.
$$

En cambio, si el fluido inferior es poco profundo,

$$
kh_2\ll 1,
\qquad
\tanh(kh_2)\simeq kh_2,
$$

entonces

$$
\omega^2\simeq gh_2k^2.
$$

**3. Inestabilidad de Rayleigh--Taylor.** Si

$$
\rho_2<\rho_1,
$$

el numerador de la relación de dispersión es negativo:

$$
(\rho_2-\rho_1)gk<0.
$$

Como el denominador es positivo, resulta

$$
\omega^2<0.
$$

Por lo tanto $\omega$ es imaginario y las perturbaciones no oscilan: crecen o decaen exponencialmente. Esto corresponde a la inestabilidad de Rayleigh--Taylor, porque el fluido más denso está por encima del menos denso.

## (c) Expansión para fluido superior no acotado y aguas poco profundas

Partimos de

$$
\omega^2
=
\frac{(\rho_2-\rho_1)gk}
{\rho_1\coth(kh_1)+\rho_2\coth(kh_2)}.
$$

En el límite de fluido superior no acotado,

$$
kh_1\to\infty,
$$

y entonces

$$
\coth(kh_1)\to 1.
$$

La relación queda

$$
\omega^2
=
\frac{(\rho_2-\rho_1)gk}
{\rho_1+\rho_2\coth(kh_2)}.
$$

Ahora definimos

$$
z=kh_2.
$$

El régimen de aguas poco profundas corresponde a

$$
z\ll 1.
$$

Usamos la expansión de Taylor

$$
\coth z
=
\frac{1}{z}
+\frac{z}{3}
-\frac{z^3}{45}
+\mathcal O(z^5).
$$

Entonces,

$$
\rho_1+\rho_2\coth z
=
\rho_1
+\rho_2
\left(
\frac{1}{z}
+\frac{z}{3}
-\frac{z^3}{45}
+\mathcal O(z^5)
\right).
$$

Sacamos factor común $\rho_2/z$:

$$
\rho_1+\rho_2\coth z
=
\frac{\rho_2}{z}
\left[
1
+\frac{\rho_1}{\rho_2}z
+\frac{z^2}{3}
-\frac{z^4}{45}
+\mathcal O(z^6)
\right].
$$

Por lo tanto,

$$
\frac{1}{\rho_1+\rho_2\coth z}
=
\frac{z}{\rho_2}
\left[
1
+\frac{\rho_1}{\rho_2}z
+\frac{z^2}{3}
+\mathcal O(z^4)
\right]^{-1}.
$$

Para obtener el primer término no lineal en la relación de dispersión, usamos

$$
(1+\varepsilon)^{-1}
=
1-\varepsilon+\mathcal O(\varepsilon^2).
$$

Con

$$
\varepsilon
=
\frac{\rho_1}{\rho_2}z
+\frac{z^2}{3}
+\cdots,
$$

al orden necesario queda

$$
\left[
1
+\frac{\rho_1}{\rho_2}z
+\frac{z^2}{3}
+\cdots
\right]^{-1}
=
1-\frac{\rho_1}{\rho_2}z
+\mathcal O(z^2).
$$

Así,

$$
\frac{1}{\rho_1+\rho_2\coth z}
\simeq
\frac{z}{\rho_2}
\left[
1-\frac{\rho_1}{\rho_2}z
\right].
$$

Volviendo a $z=kh_2$,

$$
\omega^2
\simeq
(\rho_2-\rho_1)gk
\frac{kh_2}{\rho_2}
\left[
1-\frac{\rho_1}{\rho_2}kh_2
\right].
$$

Definimos

$$
c_0^2
=
\frac{(\rho_2-\rho_1)gh_2}{\rho_2}.
$$

Entonces

$$
\boxed{
\omega^2
\simeq
c_0^2 k^2
-
c_0^2
\frac{\rho_1}{\rho_2}
h_2 k^3
}
$$

o, equivalentemente,

$$
\omega^2
\simeq
c_0^2 k^2
\left(
1-\frac{\rho_1}{\rho_2}kh_2
\right).
$$

Si queremos escribir la expansión para $\omega$, tomamos raíz:

$$
\omega
=
c_0 k
\sqrt{
1-\frac{\rho_1}{\rho_2}kh_2
}.
$$

Usando

$$
\sqrt{1+\delta}
=
1+\frac{\delta}{2}
+\mathcal O(\delta^2),
$$

con

$$
\delta
=
-\frac{\rho_1}{\rho_2}kh_2,
$$

obtenemos

$$
\boxed{
\omega(k)
\simeq
c_0 k
-
\frac{1}{2}c_0
\frac{\rho_1}{\rho_2}
h_2 k^2
}
$$

Este es el primer término dispersivo no lineal en $k$ para el régimen de aguas poco profundas.

## (d) Ausencia de resonancias de tres ondas

Una interacción resonante de tres ondas debe satisfacer simultáneamente

$$
k_1=k_2+k_3,
$$

y

$$
\omega_1=\omega_2+\omega_3,
$$

con

$$
k_1\neq k_2,
\qquad
k_1\neq k_3.
$$

Usamos la expansión obtenida en el inciso anterior:

$$
\omega(k)
\simeq
c_0 k
-\lambda k^2,
$$

donde

$$
\lambda
=
\frac{1}{2}c_0
\frac{\rho_1}{\rho_2}
h_2.
$$

Si hubiera resonancia, entonces con

$$
k_1=k_2+k_3
$$

debería cumplirse

$$
\omega(k_2+k_3)
=
\omega(k_2)+\omega(k_3).
$$

Calculamos ambos lados. Primero,

$$
\omega(k_2+k_3)
=
c_0(k_2+k_3)
-\lambda(k_2+k_3)^2.
$$

Expandiendo el cuadrado,

$$
\omega(k_2+k_3)
=
c_0k_2+c_0k_3
-\lambda k_2^2
-2\lambda k_2k_3
-\lambda k_3^2.
$$

Por otro lado,

$$
\omega(k_2)+\omega(k_3)
=
\left(c_0k_2-\lambda k_2^2\right)
+
\left(c_0k_3-\lambda k_3^2\right).
$$

Es decir,

$$
\omega(k_2)+\omega(k_3)
=
c_0k_2+c_0k_3
-\lambda k_2^2
-\lambda k_3^2.
$$

La diferencia entre ambos lados es

$$
\omega(k_2+k_3)
-
\omega(k_2)
-
\omega(k_3)
=
-2\lambda k_2k_3.
$$

Para que exista resonancia debería ser

$$
-2\lambda k_2k_3=0.
$$

Como $\lambda\neq 0$, esto requiere

$$
k_2=0
\qquad
\text{o}
\qquad
k_3=0.
$$

Pero eso corresponde a una interacción trivial, no a una resonancia de tres ondas entre modos distintos. Por lo tanto, no hay interacción resonante no trivial de tres ondas para la relación de dispersión hallada.

En palabras: el término lineal $c_0k$ por sí solo sería no dispersivo y permitiría igualdades degeneradas, pero el primer término dispersivo proporcional a $k^2$ rompe esa posibilidad para modos no nulos.

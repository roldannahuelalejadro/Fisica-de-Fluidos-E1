
## **Problema 2: Análisis de ecuaciones diferenciales lineales 2D**
 Considere los siguientes sistemas dinámicos:

1.   $\dot{x} = 4x-y$, $\dot{y} = 2x + y$
2.    $\dot{x} = 5x + 2y$, $\dot{y} = 17x - 5y$
3.   $\dot{x} = y$, $\dot{y} = -x$

Para cada sistema:

(i) Escriba al sistema en forma matricial $\mathbf{\dot{x}}=\mathbf{A}\mathbf{x}$. Busque el polinomio característico.

(ii) Encuentre la solución general del sistema

(iii) Clasifique el punto fijo (estable, inestable o saddle)

(iv) Realice el diagrama de fases en papel y compare numericamente



## Sistema 1

Tenemos el sistema

$$
\dot{x}=4x-y,
\qquad
\dot{y}=2x+y.
$$

Lo escribimos en forma matricial como

$$
\dot{\mathbf{x}}=A\mathbf{x},
$$

donde

$$
\mathbf{x}
=
\begin{pmatrix}
x\\
y
\end{pmatrix},
\qquad
A=
\begin{pmatrix}
4 & -1\\
2 & 1
\end{pmatrix}.
$$

---

### (i) Polinomio característico

Buscamos los autovalores resolviendo

$$
\det(A-\lambda I)=0.
$$

Tenemos

$$
A-\lambda I
=
\begin{pmatrix}
4-\lambda & -1\\
2 & 1-\lambda
\end{pmatrix}.
$$

Entonces

$$
\det(A-\lambda I)
=
(4-\lambda)(1-\lambda)-(-1)(2).
$$

Por lo tanto,

$$
\det(A-\lambda I)
=
(4-\lambda)(1-\lambda)+2.
$$

Desarrollando,

$$
(4-\lambda)(1-\lambda)
=
4-5\lambda+\lambda^2.
$$

Entonces,

$$
p(\lambda)
=
\lambda^2-5\lambda+6.
$$

Factorizamos:

$$
p(\lambda)
=
(\lambda-2)(\lambda-3).
$$

Por lo tanto,

$$
\lambda_1=2,
\qquad
\lambda_2=3.
$$

---

### (ii) Solución general

Para

$$
\lambda_1=2,
$$

resolvemos

$$
(A-2I)\mathbf{v}_1=0.
$$

Como

$$
A-2I
=
\begin{pmatrix}
2 & -1\\
2 & -1
\end{pmatrix},
$$

la ecuación es

$$
2x-y=0.
$$

Entonces

$$
y=2x.
$$

Podemos tomar

$$
\mathbf{v}_1=
\begin{pmatrix}
1\\
2
\end{pmatrix}.
$$

Para

$$
\lambda_2=3,
$$

resolvemos

$$
(A-3I)\mathbf{v}_2=0.
$$

Como

$$
A-3I
=
\begin{pmatrix}
1 & -1\\
2 & -2
\end{pmatrix},
$$

la ecuación es

$$
x-y=0.
$$

Entonces

$$
y=x.
$$

Podemos tomar

$$
\mathbf{v}_2=
\begin{pmatrix}
1\\
1
\end{pmatrix}.
$$

Por lo tanto, la solución general es

$$
\mathbf{x}(t)
=
C_1 e^{2t}
\begin{pmatrix}
1\\
2
\end{pmatrix}
+
C_2 e^{3t}
\begin{pmatrix}
1\\
1
\end{pmatrix}.
$$

Es decir,

$$
x(t)=C_1e^{2t}+C_2e^{3t},
$$

$$
y(t)=2C_1e^{2t}+C_2e^{3t}.
$$

---

### (iii) Clasificación del punto fijo

El único punto fijo es

$$
(x^\*,y^\*)=(0,0).
$$

Como ambos autovalores son reales y positivos,

$$
\lambda_1=2>0,
\qquad
\lambda_2=3>0,
$$

las trayectorias se alejan del origen.

Por lo tanto,

$$
\boxed{(0,0)\text{ es un nodo inestable.}}
$$

---

### (iv) Diagrama de fases

Las direcciones propias son

$$
\mathbf{v}_1=
\begin{pmatrix}
1\\
2
\end{pmatrix},
\qquad
\mathbf{v}_2=
\begin{pmatrix}
1\\
1
\end{pmatrix}.
$$

Por lo tanto, las rectas invariantes son

$$
y=2x,
\qquad
y=x.
$$

Como ambos autovalores son positivos, el flujo se aleja del origen sobre ambas direcciones.

El diagrama de fases corresponde a una fuente o nodo inestable.

# Problema 10.2 (iv) — Puntos fijos y estabilidad del sistema de dos modos

Consideramos el sistema obtenido en 10.2(iii):

$$
\begin{aligned}
\dot a_1&=(1-k^2)a_1+\frac12a_1a_2,\\
\dot a_2&=4(1-4k^2)a_2-\frac12a_1^2.
\end{aligned}
$$

Buscaremos todos sus puntos fijos y determinaremos su estabilidad en función de $k=2\pi/L$.

## 1. Ecuaciones de punto fijo

En un punto fijo se cumple

$$
\dot a_1=\dot a_2=0.
$$

Por lo tanto,

$$
0=(1-k^2)a_1+\frac12a_1a_2
=a_1\left(1-k^2+\frac{a_2}{2}\right),
$$

$$
0=4(1-4k^2)a_2-\frac12a_1^2.
$$

La primera ecuación produce dos casos:

$$
a_1=0,
$$

o bien

$$
a_2=2(k^2-1).
$$

## 2. Punto fijo trivial

Si $a_1=0$, la segunda ecuación se reduce a

$$
4(1-4k^2)a_2=0.
$$

Para $k\neq1/2$ resulta

$$
a_2=0.
$$

Así obtenemos el punto fijo trivial

$$
\boxed{P_0=(0,0)}.
$$

Existe un caso degenerado en $k=1/2$: allí $a_1=0$ y cualquier valor de $a_2$ satisface las ecuaciones de punto fijo. Por tanto, exactamente en ese valor aparece una línea de equilibrios:

$$
\boxed{k=\frac12:\quad (a_1,a_2)=(0,A),\quad A\in\mathbb R.}
$$

Este detalle no afecta la clasificación en los intervalos abiertos, pero será relevante al discutir bifurcaciones.

## 3. Puntos fijos no triviales

En el segundo caso,

$$
a_2=2(k^2-1).
$$

Sustituyendo en la segunda ecuación:

$$
4(1-4k^2)\,2(k^2-1)-\frac12a_1^2=0.
$$

Entonces

$$
a_1^2=16(k^2-1)(1-4k^2).
$$

Por lo tanto aparecen dos soluciones simétricas:

$$
\boxed{
P_\pm=
\left(
\pm4\sqrt{(k^2-1)(1-4k^2)},
\;2(k^2-1)
\right)
}.
$$

Para que sean reales necesitamos

$$
(k^2-1)(1-4k^2)\geq0.
$$

Tomando $k>0$, los puntos no triviales existen para

$$
\boxed{\frac12\leq k\leq1}.
$$

En el interior del intervalo son dos puntos distintos. Como $k=2\pi/L$, esto equivale a

$$
\boxed{2\pi\leq L\leq4\pi}.
$$

## 4. Matriz jacobiana

Definimos

$$
f_1(a_1,a_2)=(1-k^2)a_1+\frac12a_1a_2,
$$

$$
f_2(a_1,a_2)=4(1-4k^2)a_2-\frac12a_1^2.
$$

La matriz jacobiana es

$$
J(a_1,a_2)=
\begin{pmatrix}
\dfrac{\partial f_1}{\partial a_1} &
\dfrac{\partial f_1}{\partial a_2}\\[2mm]
\dfrac{\partial f_2}{\partial a_1} &
\dfrac{\partial f_2}{\partial a_2}
\end{pmatrix},
$$

es decir,

$$
\boxed{
J(a_1,a_2)=
\begin{pmatrix}
(1-k^2)+a_2/2 & a_1/2\\
-a_1 & 4(1-4k^2)
\end{pmatrix}
}.
$$

## 5. Estabilidad del punto trivial

Evaluando en $P_0=(0,0)$:

$$
J(P_0)=
\begin{pmatrix}
1-k^2 & 0\\
0 & 4(1-4k^2)
\end{pmatrix}.
$$

Los autovalores son

$$
\lambda_1=1-k^2,
\qquad
\lambda_2=4(1-4k^2).
$$

Por tanto:

- Si $k>1$, ambos autovalores son negativos y $P_0$ es estable.
- Si $1/2<k<1$, $\lambda_1>0$ y $\lambda_2<0$: $P_0$ es un punto silla.
- Si $0<k<1/2$, ambos autovalores son positivos y $P_0$ es un nodo inestable.

En términos de $L$:

$$
\boxed{
\begin{array}{c|c}
L<2\pi & P_0\text{ estable}\\
2\pi<L<4\pi & P_0\text{ silla}\\
L>4\pi & P_0\text{ inestable}
\end{array}}
$$

En el manuscrito se utiliza a veces “inestable” en sentido general para $1/2<k<1$. La clasificación más precisa en esa región es **punto silla**, porque los autovalores tienen signos opuestos.

## 6. Estabilidad de $P_\pm$

En los puntos no triviales se cumple

$$
a_2=2(k^2-1),
$$

de modo que la entrada superior izquierda del Jacobiano se anula:

$$
(1-k^2)+\frac{a_2}{2}=0.
$$

Entonces

$$
J(P_\pm)=
\begin{pmatrix}
0 & a_1^*/2\\
-a_1^* & 4(1-4k^2)
\end{pmatrix}.
$$

Su traza y determinante son

$$
\operatorname{tr}J=4(1-4k^2),
$$

$$
\det J=\frac{(a_1^*)^2}{2}
=8(k^2-1)(1-4k^2).
$$

En el intervalo de existencia $1/2<k<1$ se tiene

$$
\operatorname{tr}J<0,
\qquad
\det J>0,
$$

por lo que ambos puntos son estables.

Los autovalores pueden escribirse como

$$
\boxed{
\lambda_\pm=
2(1-4k^2)
\pm2\sqrt{3(1-4k^2)(1-2k^2)}
}.
$$

De aquí se distinguen dos comportamientos:

$$
\frac12<k<\frac{1}{\sqrt2}
\quad\Longrightarrow\quad
P_\pm\text{ son focos o espirales estables},
$$

$$
\frac{1}{\sqrt2}<k<1
\quad\Longrightarrow\quad
P_\pm\text{ son nodos estables}.
$$

En $k=1/\sqrt2$ ocurre la transición entre autovalores complejos y reales.

## 7. Resumen de estabilidad

Para $k>0$ y fuera de los valores degenerados:

| Intervalo de $k$ | Punto $P_0$ | Puntos $P_\pm$ |
|---|---|---|
| $k>1$ | estable | no existen |
| $1/\sqrt2<k<1$ | silla | nodos estables |
| $1/2<k<1/\sqrt2$ | silla | espirales estables |
| $0<k<1/2$ | nodo inestable | no existen |

Los puntos $P_+$ y $P_-$ tienen siempre la misma estabilidad porque están relacionados por la simetría

$$
(a_1,a_2)\mapsto(-a_1,a_2).
$$

## 8. Comprobación numérica

La siguiente celda evalúa los puntos fijos y los autovalores del Jacobiano para un representante de cada región. Sirve como control de las expresiones analíticas.

```python
import numpy as np


def jacobiano(a1, a2, k):
    return np.array([
        [(1-k**2) + 0.5*a2, 0.5*a1],
        [-a1, 4*(1-4*k**2)],
    ])


def puntos_fijos(k):
    puntos = [("P0", 0.0, 0.0)]
    radicando = (k**2-1)*(1-4*k**2)
    if radicando > 0:
        a1 = 4*np.sqrt(radicando)
        a2 = 2*(k**2-1)
        puntos.extend([("P+", a1, a2), ("P-", -a1, a2)])
    return puntos


for k in (1.2, 0.85, 0.60, 0.40):
    print(f"\nk = {k:.2f}")
    for nombre, a1, a2 in puntos_fijos(k):
        eig = np.linalg.eigvals(jacobiano(a1, a2, k))
        print(f"  {nombre}: ({a1:+.4f}, {a2:+.4f}), autovalores={eig}")
```

## Conclusión

El sistema posee el equilibrio trivial para todo $k$ y un par simétrico de equilibrios no triviales cuando

$$
\frac12<k<1.
$$

Al atravesar $k=1$, el punto trivial pierde estabilidad y aparecen los dos puntos estables $P_\pm$. En $k=1/2$ el truncamiento presenta una degeneración: los puntos no triviales alcanzan la línea $a_1=0$ y aparece una familia continua de equilibrios.

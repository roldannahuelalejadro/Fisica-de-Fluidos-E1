# Ejercicio 10.2(i): Expansión de Galerkin con simetría impar

Queremos obtener el sistema de ecuaciones para los modos de Fourier de la ecuación de Kuramoto-Sivashinsky reescalada:

$$
u_t=-u_{xx}-k^2u_{xxxx}-u u_x.
$$

Trabajamos en el dominio

$$
0\leq x\leq 2\pi,
$$

y usamos una expansión impar, es decir, solo con modos seno:

$$
u(x,t)=\sum_{\ell=1}^{\infty} a_\ell(t)\sin(\ell x).
$$

El objetivo es llegar a una ecuación para cada coeficiente modal $a_q(t)$, usando $q$ como índice del modo sobre el que proyectamos.

> Nota: en el desarrollo uso $\ell$ y $m$ como índices cruzados del producto $u u_x$, y $q$ como índice del modo de prueba $\sin(qx)$.

## 1. Expansión de $u$ y sus derivadas

Partimos de

$$
u(x,t)=\sum_{\ell=1}^{\infty} a_\ell(t)\sin(\ell x).
$$

Como los coeficientes dependen del tiempo,

$$
a_\ell=a_\ell(t),
$$

se tiene

$$
u_t(x,t)=\sum_{\ell=1}^{\infty}\dot a_\ell(t)\sin(\ell x).
$$

Las derivadas espaciales son

$$
u_x(x,t)=\sum_{\ell=1}^{\infty}\ell a_\ell(t)\cos(\ell x),
$$

$$
u_{xx}(x,t)=-\sum_{\ell=1}^{\infty}\ell^2 a_\ell(t)\sin(\ell x),
$$

y

$$
u_{xxxx}(x,t)=\sum_{\ell=1}^{\infty}\ell^4 a_\ell(t)\sin(\ell x).
$$

## 2. Sustitución de los términos lineales

La ecuación es

$$
u_t=-u_{xx}-k^2u_{xxxx}-u u_x.
$$

Los términos lineales quedan:

$$
-u_{xx}=\sum_{\ell=1}^{\infty}\ell^2 a_\ell(t)\sin(\ell x),
$$

y

$$
-k^2u_{xxxx}=-\sum_{\ell=1}^{\infty}k^2\ell^4a_\ell(t)\sin(\ell x).
$$

Por lo tanto,

$$
-u_{xx}-k^2u_{xxxx}=\sum_{\ell=1}^{\infty}\left(\ell^2-k^2\ell^4\right)a_\ell(t)\sin(\ell x).
$$

También podemos escribir

$$
\ell^2-k^2\ell^4=(1-k^2\ell^2)\ell^2.
$$

## 3. Término no lineal $u u_x$

Para el término no lineal hay que tener cuidado porque aparece un producto de dos series. Usamos dos índices cruzados: $\ell$ para la serie de $u$ y $m$ para la serie de $u_x$.

Tenemos

$$
u(x,t)=\sum_{\ell=1}^{\infty}a_\ell(t)\sin(\ell x),
$$

y

$$
u_x(x,t)=\sum_{m=1}^{\infty}m a_m(t)\cos(mx).
$$

Entonces

$$
u u_x=\left(\sum_{\ell=1}^{\infty}a_\ell(t)\sin(\ell x)\right)\left(\sum_{m=1}^{\infty}m a_m(t)\cos(mx)\right).
$$

Por lo tanto,

$$
u u_x=\sum_{\ell=1}^{\infty}\sum_{m=1}^{\infty}m a_\ell(t)a_m(t)\sin(\ell x)\cos(mx).
$$

Usamos la identidad trigonométrica

$$
\sin(\ell x)\cos(mx)=\frac{1}{2}\left[\sin((\ell+m)x)+\sin((\ell-m)x)\right].
$$

Entonces

$$
u u_x=\frac{1}{2}\sum_{\ell=1}^{\infty}\sum_{m=1}^{\infty}m a_\ell(t)a_m(t)\left[\sin((\ell+m)x)+\sin((\ell-m)x)\right].
$$

## 4. Proyección de Galerkin sobre $\sin(qx)$

Para obtener la ecuación del modo $q$, multiplicamos la ecuación por $\sin(qx)$ e integramos en un período:

$$
\int_0^{2\pi}(\cdots)\sin(qx)\,dx.
$$

Usamos la ortogonalidad

$$
\int_0^{2\pi}\sin(nx)\sin(qx)\,dx=\pi\delta_{nq},\qquad n,q\geq 1.
$$

La parte temporal da

$$
\int_0^{2\pi}u_t\sin(qx)\,dx
=\int_0^{2\pi}\sum_{\ell=1}^{\infty}\dot a_\ell(t)\sin(\ell x)\sin(qx)\,dx.
$$

Por ortogonalidad,

$$
\int_0^{2\pi}u_t\sin(qx)\,dx=\pi\dot a_q(t).
$$

La parte lineal da

$$
\int_0^{2\pi}\left[-u_{xx}-k^2u_{xxxx}\right]\sin(qx)\,dx=\pi(q^2-k^2q^4)a_q(t).
$$

Por lo tanto, antes de simplificar el término no lineal,

$$
\pi\dot a_q(t)=\pi(q^2-k^2q^4)a_q(t)-\int_0^{2\pi}u u_x\sin(qx)\,dx.
$$

Dividiendo por $\pi$,

$$
\dot a_q(t)=(q^2-k^2q^4)a_q(t)-\frac{1}{\pi}\int_0^{2\pi}u u_x\sin(qx)\,dx.
$$

## 5. Proyección del término no lineal

Partimos de

$$
u u_x=\frac{1}{2}\sum_{\ell=1}^{\infty}\sum_{m=1}^{\infty}m a_\ell(t)a_m(t)\left[\sin((\ell+m)x)+\sin((\ell-m)x)\right].
$$

Entonces

$$
\int_0^{2\pi}u u_x\sin(qx)\,dx
=\frac{1}{2}\sum_{\ell=1}^{\infty}\sum_{m=1}^{\infty}m a_\ell(t)a_m(t)
\int_0^{2\pi}\left[\sin((\ell+m)x)+\sin((\ell-m)x)\right]\sin(qx)\,dx.
$$

Ahora analizamos qué combinaciones sobreviven por ortogonalidad.

### 5.1. Aporte de $\sin((\ell+m)x)$

El término

$$
\sin((\ell+m)x)
$$

contribuye cuando

$$
\ell+m=q.
$$

Entonces

$$
\ell=q-m.
$$

Como $\ell\geq 1$, este aporte existe para

$$
m=1,\dots,q-1.
$$

Por lo tanto,

$$
\frac{1}{2}\sum_{\ell=1}^{\infty}\sum_{m=1}^{\infty}m a_\ell a_m
\int_0^{2\pi}\sin((\ell+m)x)\sin(qx)\,dx
=\frac{\pi}{2}\sum_{m=1}^{q-1}m a_{q-m}a_m.
$$

### 5.2. Aporte de $\sin((\ell-m)x)$

Ahora miramos

$$
\sin((\ell-m)x).
$$

Hay dos casos relevantes.

#### Caso A: $\ell-m=q$

Si

$$
\ell-m=q,
$$

entonces

$$
\ell=q+m.
$$

Este caso aporta con signo positivo:

$$
\frac{\pi}{2}\sum_{m=1}^{\infty}m a_{q+m}a_m.
$$

#### Caso B: $\ell-m=-q$

Si

$$
\ell-m=-q,
$$

entonces

$$
\ell=m-q.
$$

Pero en este caso

$$
\sin((\ell-m)x)=\sin(-qx)=-\sin(qx).
$$

Por eso aparece un signo negativo. Este caso existe cuando

$$
m>q.
$$

Su aporte es

$$
-\frac{\pi}{2}\sum_{m=q+1}^{\infty}m a_{m-q}a_m.
$$

## 6. Escritura compacta con función signo

Los dos aportes de diferencia de índices se pueden juntar usando

$$
\operatorname{sg}(q-m)a_{|q-m|}.
$$

La convención es

$$
\operatorname{sg}(q-m)a_{|q-m|}=
\begin{cases}
+a_{q-m}, & q>m,\\
0, & q=m,\\
-a_{m-q}, & q<m.
\end{cases}
$$

Entonces

$$
\frac{\pi}{2}\sum_{m=1}^{q-1}m a_{q-m}a_m
-
\frac{\pi}{2}\sum_{m=q+1}^{\infty}m a_{m-q}a_m
=
\frac{\pi}{2}\sum_{m=1}^{\infty}m a_m\operatorname{sg}(q-m)a_{|q-m|}.
$$

El aporte del caso $\ell=q+m$ era

$$
\frac{\pi}{2}\sum_{m=1}^{\infty}m a_{q+m}a_m.
$$

Juntando todo,

$$
\int_0^{2\pi}u u_x\sin(qx)\,dx
=\frac{\pi}{2}\sum_{m=1}^{\infty}m a_m(t)
\left[\operatorname{sg}(q-m)a_{|q-m|}(t)+a_{q+m}(t)\right].
$$

Por lo tanto,

$$
\frac{1}{\pi}\int_0^{2\pi}u u_x\sin(qx)\,dx
=\frac{1}{2}\sum_{m=1}^{\infty}m a_m(t)
\left[\operatorname{sg}(q-m)a_{|q-m|}(t)+a_{q+m}(t)\right].
$$

## 7. Sistema final para los modos

Recordemos que

$$
\dot a_q(t)=(q^2-k^2q^4)a_q(t)-\frac{1}{\pi}\int_0^{2\pi}u u_x\sin(qx)\,dx.
$$

Sustituyendo el resultado anterior,

$$
\dot a_q(t)
=(q^2-k^2q^4)a_q(t)
-
\frac{1}{2}\sum_{m=1}^{\infty}m a_m(t)
\left[\operatorname{sg}(q-m)a_{|q-m|}(t)+a_{q+m}(t)\right].
$$

Como

$$
q^2-k^2q^4=(1-k^2q^2)q^2,
$$

también podemos escribir

$$
\boxed{
\dot a_q(t)
=
(1-k^2q^2)q^2a_q(t)
-
\frac{1}{2}\sum_{m=1}^{\infty}m a_m(t)
\left[\operatorname{sg}(q-m)a_{|q-m|}(t)+a_{q+m}(t)\right]
}
$$

para

$$
q=1,2,3,\dots
$$

Esta es la expresión pedida para la expansión de Galerkin con simetría impar.

Si se trunca a $N$ modos, se toma

$$
q=1,\dots,N,
$$

y se considera

$$
a_j=0
$$

para los índices fuera del rango retenido, por ejemplo $j>N$.

## 8. Comprobación numérica opcional de la integral no lineal

La siguiente celda no es necesaria para la demostración, pero sirve como control numérico simple. Calcula

$$
\int_0^{2\pi}u u_x\sin(qx)\,dx
$$

de dos maneras:

1. por integración numérica directa;
2. usando la fórmula compacta con $\operatorname{sg}(q-m)$.

Para un truncamiento finito deberían coincidir aproximadamente.

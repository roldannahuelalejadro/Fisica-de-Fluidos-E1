# Problema 10.2: análisis lineal del sistema modal

En este cuaderno pasamos a limpio la resolución manuscrita correspondiente al análisis del punto fijo trivial de la expansión de Galerkin.

Partimos del sistema modal obtenido en el inciso anterior:

$$
\dot a_\ell(t)
=
(1-k^2\ell^2)\ell^2 a_\ell(t)
-
\frac{1}{2}
\sum_{m=1}^{\infty}
m a_m(t)
\left[
\operatorname{sg}(\ell-m)a_{|\ell-m|}(t)
+
a_{\ell+m}(t)
\right].
$$

Aquí

$$
\ell=1,2,3,\dots
$$

es el índice del modo, y

$$
k=\frac{2\pi}{L}.
$$

## 1. Punto fijo trivial

El sistema tiene como punto fijo trivial

$$
a_\ell^*=0
\qquad
\forall \ell.
$$

En efecto, si

$$
a_\ell(t)=0
\qquad
\forall \ell,
$$

entonces el término lineal se anula:

$$
(1-k^2\ell^2)\ell^2 a_\ell(t)=0,
$$

y el término no lineal también se anula, porque todos sus términos contienen productos de coeficientes:

$$
a_m(t)a_{|\ell-m|}(t),
\qquad
a_m(t)a_{\ell+m}(t).
$$

Por lo tanto,

$$
\dot a_\ell(t)=0
\qquad
\forall \ell,
$$

y entonces

$$
\boxed{
a_\ell^*=0
\quad
\forall \ell
}
$$

es un punto fijo del sistema modal.

## 2. Perturbación alrededor del punto fijo

Para estudiar la estabilidad del punto fijo trivial, perturbamos cada modo:

$$
a_\ell(t)=a_\ell^*+\varepsilon_\ell(t).
$$

Como

$$
a_\ell^*=0,
$$

esto queda simplemente como

$$
a_\ell(t)=\varepsilon_\ell(t).
$$

Reemplazamos en el sistema modal:

$$
\dot\varepsilon_\ell(t)
=
(1-k^2\ell^2)\ell^2 \varepsilon_\ell(t)
-
\frac{1}{2}
\sum_{m=1}^{\infty}
m \varepsilon_m(t)
\left[
\operatorname{sg}(\ell-m)\varepsilon_{|\ell-m|}(t)
+
\varepsilon_{\ell+m}(t)
\right].
$$

## 3. Linealización

La parte no lineal contiene productos de perturbaciones:

$$
\varepsilon_m(t)\varepsilon_{|\ell-m|}(t),
\qquad
\varepsilon_m(t)\varepsilon_{\ell+m}(t).
$$

Por lo tanto, esos términos son de orden cuadrático:

$$
O(\varepsilon^2).
$$

En la linealización alrededor del punto fijo se conservan solamente los términos de primer orden en \(\varepsilon\). Por eso descartamos el término cuadrático, y queda:

$$
\boxed{
\dot\varepsilon_\ell(t)
=
(1-k^2\ell^2)\ell^2 \varepsilon_\ell(t)
}
$$

para cada modo

$$
\ell=1,2,3,\dots
$$

Es decir, a primer orden, los modos quedan desacoplados.

## 4. Solución del sistema linealizado

La ecuación linealizada para cada modo es

$$
\dot\varepsilon_\ell(t)
=
\lambda_\ell \varepsilon_\ell(t),
$$

donde

$$
\lambda_\ell
=
(1-k^2\ell^2)\ell^2.
$$

La solución es

$$
\varepsilon_\ell(t)
=
\varepsilon_\ell(0)e^{\lambda_\ell t}.
$$

Por lo tanto,

$$
\boxed{
\varepsilon_\ell(t)
=
\varepsilon_\ell(0)
e^{(1-k^2\ell^2)\ell^2t}
}
$$

## 5. Condición de estabilidad de cada modo

La estabilidad del modo $\ell$ depende del signo de

$$
\lambda_\ell=(1-k^2\ell^2)\ell^2.
$$

Como

$$
\ell^2>0,
$$

el signo de \(\lambda_\ell\) está determinado por

$$
1-k^2\ell^2.
$$

Entonces el modo \(\ell\) es estable si

$$
\lambda_\ell<0.
$$

Esto ocurre cuando

$$
(1-k^2\ell^2)\ell^2<0.
$$

Como \(\ell^2>0\), esto equivale a

$$
1-k^2\ell^2<0.
$$

Por lo tanto,

$$
k^2\ell^2>1.
$$

Tomando \(k>0\) y \(\ell>0\), queda

$$
k\ell>1.
$$

Entonces:

$$
\boxed{
\text{el modo }\ell\text{ es estable si }k\ell>1.
}
$$

De forma análoga, el modo \(\ell\) es inestable si

$$
k\ell<1.
$$

## 6. Condición en términos del tamaño del dominio \(L\)

Recordemos que

$$
k=\frac{2\pi}{L}.
$$

La condición de inestabilidad de un modo es

$$
k\ell<1.
$$

Reemplazando \(k\):

$$
\frac{2\pi}{L}\ell<1.
$$

Entonces

$$
\frac{2\pi\ell}{L}<1.
$$

Multiplicando por \(L>0\):

$$
2\pi\ell<L.
$$

Por lo tanto,

$$
\boxed{
L>2\pi\ell
}
$$

es la condición para que el modo \(\ell\) sea inestable.

En particular, el primer modo \(\ell=1\) se vuelve inestable cuando

$$
\boxed{
L>2\pi.
}
$$

## 7. Estabilidad del punto fijo trivial

Para que el punto fijo trivial

$$
a_\ell^*=0
\qquad
\forall \ell
$$

sea linealmente estable, necesitamos que todos los modos sean estables. Es decir,

$$
k\ell>1
\qquad
\forall \ell\geq 1.
$$

El modo más restrictivo es \(\ell=1\), porque si

$$
k>1,
$$

entonces automáticamente

$$
k\ell>1
$$

para todo \(\ell\geq1\).

Por lo tanto, la condición de estabilidad del punto fijo trivial es

$$
k>1.
$$

Como

$$
k=\frac{2\pi}{L},
$$

esto equivale a

$$
\frac{2\pi}{L}>1.
$$

Entonces

$$
2\pi>L.
$$

Por lo tanto,

$$
\boxed{
L<2\pi
\Rightarrow
a_\ell^*=0
\text{ es linealmente estable.}
}
$$

Mientras que si

$$
L>2\pi,
$$

el primer modo se vuelve inestable y el punto fijo trivial pierde estabilidad:

$$
\boxed{
L>2\pi
\Rightarrow
a_\ell^*=0
\text{ es linealmente inestable.}
}
$$

## 8. Resumen

El sistema modal es

$$
\dot a_\ell(t)
=
(1-k^2\ell^2)\ell^2 a_\ell(t)
-
\frac{1}{2}
\sum_{m=1}^{\infty}
m a_m(t)
\left[
\operatorname{sg}(\ell-m)a_{|\ell-m|}(t)
+
a_{\ell+m}(t)
\right].
$$

El punto fijo trivial es

$$
a_\ell^*=0
\qquad
\forall \ell.
$$

Al linealizar alrededor de este punto fijo, el término no lineal desaparece por ser cuadrático en los coeficientes \(a_m\). Queda

$$
\dot a_\ell(t)
=
(1-k^2\ell^2)\ell^2 a_\ell(t).
$$

El autovalor del modo \(\ell\) es

$$
\lambda_\ell=(1-k^2\ell^2)\ell^2.
$$

Por lo tanto:

$$
\lambda_\ell<0
\iff
k\ell>1,
$$

y

$$
\lambda_\ell>0
\iff
k\ell<1.
$$

Como

$$
k=\frac{2\pi}{L},
$$

el modo \(\ell\) es inestable cuando

$$
L>2\pi\ell.
$$

En particular, el primer modo pierde estabilidad cuando

$$
L>2\pi.
$$

Por lo tanto,

$$
\boxed{
L<2\pi
\Rightarrow
\text{el punto fijo trivial es linealmente estable}
}
$$

y

$$
\boxed{
L>2\pi
\Rightarrow
\text{el punto fijo trivial es linealmente inestable.}
}
$$

## 9. Comprobación numérica simple de los autovalores

La siguiente celda grafica los autovalores lineales

$$
\lambda_\ell=(1-k^2\ell^2)\ell^2
$$

para distintos valores de \(L\). Esto permite ver cuándo aparecen modos inestables.

```python
import numpy as np
import matplotlib.pyplot as plt

def lambdas_lineales(L, N):
    k = 2*np.pi/L
    ell = np.arange(1, N+1)
    lam = (1 - k**2 * ell**2) * ell**2
    return ell, lam

N = 12
Ls = [1.5*np.pi, 2*np.pi, 3*np.pi, 6*np.pi]

plt.figure(figsize=(8, 5))

for L in Ls:
    ell, lam = lambdas_lineales(L, N)
    plt.plot(ell, lam, marker="o", label=rf"$L={L/np.pi:.1f}\pi$")

plt.axhline(0, color="black", linewidth=1)
plt.xlabel(r"modo $\ell$")
plt.ylabel(r"$\lambda_\ell$")
plt.title(r"Autovalores lineales $\lambda_\ell=(1-k^2\ell^2)\ell^2$")
plt.legend()
plt.grid(False)
plt.show()
```

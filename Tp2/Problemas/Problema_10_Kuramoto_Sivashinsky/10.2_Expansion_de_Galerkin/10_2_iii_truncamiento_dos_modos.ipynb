# Problema 10.2 (iii) — Truncamiento de Galerkin a dos modos

Partimos del sistema modal obtenido en 10.2(i):

$$
\dot a_\ell
=(1-k^2\ell^2)\ell^2a_\ell
-\frac{1}{2}\sum_{m=1}^{\infty}m a_m
\left[\operatorname{sg}(\ell-m)a_{|\ell-m|}+a_{\ell+m}\right].
$$

Ahora retenemos solamente los dos primeros modos:

$$
u(x,t)\simeq a_1(t)\sin x+a_2(t)\sin(2x).
$$

Esto significa que

$$
a_0=0,
\qquad
a_m=0\quad\text{para }m>2.
$$

El objetivo es escribir un sistema cerrado de dos ecuaciones para $a_1(t)$ y $a_2(t)$, siguiendo la notación de los manuscritos.

## 1. Ecuación del primer modo

Tomamos $\ell=1$. La parte lineal es

$$
(1-k^2)a_1.
$$

En la suma no lineal solo debemos considerar $m=1,2$:

$$
S_1=-\frac12\sum_{m=1}^{2}m a_m
\left[\operatorname{sg}(1-m)a_{|1-m|}+a_{1+m}\right].
$$

Para $m=1$:

$$
-\frac12a_1\left[\operatorname{sg}(0)a_0+a_2\right]
=-\frac12a_1a_2.
$$

Para $m=2$:

$$
-\frac12(2a_2)\left[\operatorname{sg}(-1)a_1+a_3\right]
=-a_2(-a_1)=a_1a_2.
$$

Por lo tanto,

$$
S_1=-\frac12a_1a_2+a_1a_2
=\frac12a_1a_2.
$$

La primera ecuación queda

$$
\boxed{
\dot a_1=(1-k^2)a_1+\frac12a_1a_2
}.
$$

## 2. Ecuación del segundo modo

Para $\ell=2$, la parte lineal es

$$
(1-4k^2)4a_2=4(1-4k^2)a_2.
$$

El término no lineal es

$$
S_2=-\frac12\sum_{m=1}^{2}m a_m
\left[\operatorname{sg}(2-m)a_{|2-m|}+a_{2+m}\right].
$$

Para $m=1$:

$$
-\frac12a_1\left[\operatorname{sg}(1)a_1+a_3\right]
=-\frac12a_1^2.
$$

Para $m=2$ aparecen únicamente $a_0$ y $a_4$, ambos nulos en este truncamiento:

$$
-\frac12(2a_2)\left[\operatorname{sg}(0)a_0+a_4\right]=0.
$$

Entonces

$$
\boxed{
\dot a_2=4(1-4k^2)a_2-\frac12a_1^2
}.
$$

## 3. Sistema truncado

El sistema de Galerkin a dos modos es

$$
\boxed{
\begin{aligned}
\dot a_1&=(1-k^2)a_1+\frac12a_1a_2,\\
\dot a_2&=4(1-4k^2)a_2-\frac12a_1^2.
\end{aligned}
}
$$

También puede escribirse como

$$
\dot{\mathbf a}=F(\mathbf a;k),
\qquad
\mathbf a=
\begin{pmatrix}a_1\\a_2\end{pmatrix}.
$$

El sistema es invariante bajo

$$
(a_1,a_2)\longmapsto(-a_1,a_2).
$$

Esta simetría anticipa que los puntos fijos no triviales aparecerán en pares con valores opuestos de $a_1$.

## 4. Comprobación directa desde el campo truncado

Podemos verificar el resultado sin utilizar la fórmula general. Si

$$
u=a_1\sin x+a_2\sin(2x),
$$

entonces

$$
u_x=a_1\cos x+2a_2\cos(2x).
$$

Al expandir $-uu_x$ y proyectar sobre $\sin x$ y $\sin(2x)$ se obtienen respectivamente

$$
\frac12a_1a_2,
\qquad
-\frac12a_1^2,
$$

que son exactamente los términos no lineales del sistema anterior.

```python
import numpy as np


def rhs_dos_modos(a1, a2, k):
    da1 = (1-k**2)*a1 + 0.5*a1*a2
    da2 = 4*(1-4*k**2)*a2 - 0.5*a1**2
    return np.array([da1, da2])


# Ejemplo de evaluación del campo vectorial.
k = 0.8
a1, a2 = 1.0, -0.5
print("F(a1,a2;k) =", rhs_dos_modos(a1, a2, k))
```

## Conclusión

El truncamiento conserva la competencia esencial de Kuramoto–Sivashinsky:

- los términos lineales determinan qué modos crecen o decaen;
- el producto no lineal acopla $a_1$ con $a_2$;
- el primer modo fuerza al segundo mediante $-a_1^2/2$;
- el segundo modo realimenta al primero mediante $a_1a_2/2$.
